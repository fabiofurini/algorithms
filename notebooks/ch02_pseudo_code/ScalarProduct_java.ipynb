## ScalarProduct

**Input:** two arrays $a[1..n]$ and $b[1..n]$ of size $n \ge 1$. **Output:** the scalar product $\sum_{i=1}^{n} a[i] \cdot b[i]$ of the two arrays.

```python
%%writefile ScalarProduct.java
import java.util.Arrays;

public class ScalarProduct {

    // Returns the scalar product of a[0..n-1] and b[0..n-1], with n >= 1
    static int scalarProduct(int[] a, int[] b, int n, boolean trace) {
        int prod = a[0] * b[0];
        if (trace)
            System.out.printf("%3s  %4s  %4s  %9s  %5d\n", "--", "--", "--", "--", prod);
        for (int i = 1; i < n; i++) {   // i = 2..n in the pseudo-code
            prod = prod + a[i] * b[i];
            if (trace)
                System.out.printf("%3d  %4d  %4d  %9d  %5d\n", i, a[i], b[i], a[i] * b[i], prod);
        }
        return prod;
    }

    static void run(String title, int[] a, int[] b) {
        int n = a.length;
        System.out.printf("%s: a = %s, b = %s, n = %d\n", title, Arrays.toString(a), Arrays.toString(b), n);
        System.out.printf("%3s  %4s  %4s  %9s  %5s\n", "i", "a[i]", "b[i]", "a[i]*b[i]", "prod");
        int result = scalarProduct(a, b, n, true);
        System.out.printf("Return: prod = %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 1) {          // “Your turn”: the numbers come from the command line (half for a, half for b)
            int n = args.length / 2;
            int[] a = new int[n], b = new int[n];
            for (int k = 0; k < n; k++) {
                a[k] = Integer.parseInt(args[k]);
                b[k] = Integer.parseInt(args[n + k]);
            }
            run("Your turn", a, b);
            return;
        }
        run("Instance 1", new int[]{1, 2, 3}, new int[]{4, 5, 6});
        run("Instance 2", new int[]{2, 0, -1}, new int[]{3, 4, 5});
    }
}
```
