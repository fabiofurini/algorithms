## InsertionSort

**Input:** an array $a[1..n]$ of $n$ natural numbers. **Output:** the array $a$ sorted in non-decreasing order.

At step $i$ the sub-array $a[1..i-1]$ is sorted: the key $a[i]$ is inserted in its place, after shifting one position to the right the greater keys. The function also returns the counters used in the lecture notes: the number of key comparisons and the number of shifts (no swaps: a shift is a single assignment).

```python
def row(i, j, key, a, action):          # one row of the trace
    print("%3s %3s %4s  %-20s  %s" % (i, j, key, a, action))


def insertion_sort(a, n, trace=False):
    comparisons = 0                     # key comparisons
    shifts = 0                          # shifts
    if trace:
        row("-", "-", "-", a, "initial state")
    i = 1                               # i = 2 in the pseudo-code
    while i < n:                        # i <= n in the pseudo-code
        key = a[i]                      # element to insert
        j = i - 1
        if trace:
            row(i, j, key, a, "key <- a[%d]=%d; j <- %d" % (i, key, j))
        while j >= 0 and a[j] > key:    # j >= 1 in the pseudo-code
            comparisons += 1
            shifts += 1
            a[j + 1] = a[j]             # shift the element to the right
            j = j - 1
            if trace:
                row("", j, key, a, "a[%d]=%d>%d: shift a[%d] <- %d; j <- %d" % (j + 1, a[j + 1], key, j + 2, a[j + 1], j))
        if j >= 0:
            comparisons += 1            # the comparison that stopped the loop (a[j] <= key)
        a[j + 1] = key                  # insert key in its position
        if trace:
            if j < 0:
                row("", "-", key, a, "j=%d: exit; insert a[%d] <- %d" % (j, j + 1, key))
            else:
                row("", "-", key, a, "a[%d]=%d<=%d: exit; insert a[%d] <- %d" % (j, a[j], key, j + 1, key))
        i = i + 1
    return comparisons, shifts
```

### Execution on the examples of the chapter

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s %3s %4s  %-20s  %s" % ("i", "j", "key", "a", "action"))
    comparisons, shifts = insertion_sort(a, n, trace=True)
    print("Sorted array: %s" % a)
    print("Key comparisons: %d, shifts: %d" % (comparisons, shifts))
    print()


run("Instance 1", [5, 2, 4, 1, 3])
run("Instance 2", [12, 11, 13, 5, 6])
```
