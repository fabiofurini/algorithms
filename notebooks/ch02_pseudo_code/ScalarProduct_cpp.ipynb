## ScalarProduct

**Input:** two arrays $a[1..n]$ and $b[1..n]$ of size $n \ge 1$. **Output:** the scalar product $\sum_{i=1}^{n} a[i] \cdot b[i]$ of the two arrays.

```python
%%writefile scalar_product.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Returns the scalar product of a[0..n-1] and b[0..n-1], with n >= 1
int ScalarProduct(int n, int *a, int *b, bool trace = false)
{
    int prod = a[0] * b[0];
    if (trace)
        printf("%3s  %4s  %4s  %9s  %5d\n", "--", "--", "--", "--", prod);
    for (int i = 1; i < n; i++)     // i = 2..n in the pseudo-code
    {
        prod = prod + a[i] * b[i];
        if (trace)
            printf("%3d  %4d  %4d  %9d  %5d\n", i, a[i], b[i], a[i] * b[i], prod);
    }
    return prod;
}

void Run(string title, vector<int> a, vector<int> b)
{
    int n = a.size();
    printf("%s: a = %s, b = %s, n = %d\n", title.c_str(), Str(n, a.data()).c_str(), Str(n, b.data()).c_str(), n);
    printf("%3s  %4s  %4s  %9s  %5s\n", "i", "a[i]", "b[i]", "a[i]*b[i]", "prod");
    int result = ScalarProduct(n, a.data(), b.data(), true);
    printf("Return: prod = %d\n\n", result);
}

int main(int argc, char *argv[])
{
    if (argc > 2)   // “Your turn”: the numbers come from the command line (half for a, half for b)
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        int n = v.size() / 2;
        Run("Your turn", vector<int>(v.begin(), v.begin() + n), vector<int>(v.begin() + n, v.begin() + 2 * n));
        return 0;
    }
    Run("Instance 1", {1, 2, 3}, {4, 5, 6});
    Run("Instance 2", {2, 0, -1}, {3, 4, 5});
    return 0;
}
```
