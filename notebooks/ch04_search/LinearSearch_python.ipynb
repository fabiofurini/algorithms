## LinearSearch

**Input:** an array $a[1..n]$ of $n$ natural numbers and a target value $m$. **Output:** the index of $m$ in $a$, or $-1$ if $m$ is not found.

The algorithm compares $m$ with the elements of the array one after the other and stops at the first one equal to $m$: the returned index is the smallest $k$ with $a[k]=m$. In the code the indices start from 0, so the returned index is the one of the pseudo-code minus 1.

```python
def linear_search(a, n, m, trace=False):
    i = 0                               # i = 1 in the pseudo-code
    while i < n:                        # i <= n in the pseudo-code
        if a[i] == m:
            if trace:
                print("%10d %3d  %-16s  %s" % (i + 1, i, "a[%d] = %d = m" % (i, a[i]), "Found! Return %d" % i))
            return i                    # element found at index i
        if trace:
            print("%10d %3d  %-16s  %s" % (i + 1, i, "a[%d] = %d != m" % (i, a[i]), "continue"))
        i = i + 1
    if trace:
        print("%10s %3d  %-16s  %s" % ("-", i, "-", "i = n: the loop ends, not found! Return -1"))
    return -1                           # element not found
```

### Execution on the examples of the chapter

```python
def run(a, m):
    n = len(a)
    print("Search: m = %d, a = %s, n = %d" % (m, a, n))
    print("%10s %3s  %-16s  %s" % ("iteration", "i", "comparison", "action"))
    result = linear_search(a, n, m, trace=True)
    print("Return: %d" % result)
    print()


a = [4, 2, 7, 1, 5, 9, 3]
run(a, 5)
run(a, 6)
run([1, 2, 2, 2, 2, 2, 3], 2)    # m appears several times: returns the first index
```
