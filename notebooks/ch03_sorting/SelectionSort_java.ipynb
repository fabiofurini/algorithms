## SelectionSort

**Input:** an array $a[1..n]$ of $n$ natural numbers. **Output:** the array $a$ sorted in non-decreasing order.

At step $i$ the inner loop looks for the index $\mathit{minIdx}$ of the minimum of $a[i..n]$ (the comparison is strict: in case of ties the first minimum wins) and then, if $\mathit{minIdx} \ne i$, swaps $a[i]$ with $a[\mathit{minIdx}]$: at most one swap per step. The function returns the number of key comparisons and the number of swaps.

```python
%%writefile SelectionSort.java
import java.util.Arrays;

public class SelectionSort {

    // One row of the trace
    static void row(String i, String j, String minIdx, int[] a, String action) {
        System.out.printf("%3s %3s %6s  %-20s  %s\n", i, j, minIdx, Arrays.toString(a), action);
    }

    // Sorts a[0..n-1] in place; returns {key comparisons, swaps}
    static int[] selectionSort(int[] a, int n, boolean trace) {
        int comparisons = 0;                // key comparisons
        int swaps = 0;                      // swaps
        if (trace)
            row("-", "-", "-", a, "initial state");

        for (int i = 0; i < n - 1; i++) {   // i = 1..n-1 in the pseudo-code
            int minIdx = i;                 // index of the minimum in a[i..n]
            if (trace)
                row("" + i, "-", "" + minIdx, a, String.format("minIdx := %d (a[%d]=%d)", i, i, a[i]));
            for (int j = i + 1; j < n; j++) {   // j = i+1..n in the pseudo-code
                comparisons++;
                if (a[j] < a[minIdx]) {
                    int old = minIdx;       // only for the trace
                    minIdx = j;             // new minimum found
                    if (trace)
                        row("", "" + j, "" + minIdx, a, String.format("a[%d]=%d<a[%d]=%d: minIdx <- %d", j, a[j], old, a[old], j));
                } else if (trace)
                    row("", "" + j, "" + minIdx, a, String.format("a[%d]=%d>=a[%d]=%d: no update", j, a[j], minIdx, a[minIdx]));
            }
            if (minIdx != i) {
                int tmp = a[i];             // swap a[i] and a[minIdx]
                a[i] = a[minIdx];
                a[minIdx] = tmp;
                swaps++;
                if (trace)
                    row("", "-", "-", a, String.format("minIdx=%d != %d: swap a[%d] <-> a[%d]", minIdx, i, i, minIdx));
            } else if (trace)
                row("", "-", "-", a, String.format("minIdx=%d: no swap", minIdx));
        }
        return new int[]{comparisons, swaps};
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s %3s %6s  %-20s  %s\n", "i", "j", "minIdx", "a", "action");
        int[] c = selectionSort(a, n, true);
        System.out.printf("Sorted array: %s\n", Arrays.toString(a));
        System.out.printf("Key comparisons: %d, swaps: %d\n\n", c[0], c[1]);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // “Your turn”: the array comes from the command line
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Your turn", v);
            return;
        }
        run("Instance 1", new int[]{5, 2, 4, 1, 3});
        run("Instance 2", new int[]{12, 11, 13, 5, 6});
    }
}
```
