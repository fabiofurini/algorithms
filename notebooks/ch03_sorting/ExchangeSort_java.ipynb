## ExchangeSort

**Input:** an array $a[1..n]$ of $n$ natural numbers. **Output:** the array $a$ sorted in non-decreasing order.

At step $i$ the element $a[i]$ is compared with every $a[j]$ with $j > i$, and the two elements are swapped whenever $a[i] > a[j]$: at the end of the step $a[i]$ is the minimum of $a[i..n]$. The function returns the number of key comparisons and the number of swaps.

```python
%%writefile ExchangeSort.java
import java.util.Arrays;

public class ExchangeSort {

    // One row of the trace
    static void row(String i, String j, int[] a, String action) {
        System.out.printf("%3s %3s  %-20s  %s\n", i, j, Arrays.toString(a), action);
    }

    // Sorts a[0..n-1] in place; returns {key comparisons, swaps}
    static int[] exchangeSort(int[] a, int n, boolean trace) {
        int comparisons = 0;                // key comparisons
        int swaps = 0;                      // swaps
        if (trace)
            row("-", "-", a, "initial state");

        for (int i = 0; i < n - 1; i++) {       // i = 1..n-1 in the pseudo-code
            for (int j = i + 1; j < n; j++) {   // j = i+1..n in the pseudo-code
                comparisons++;
                int ai = a[i], aj = a[j];       // values before the swap, for the trace
                if (a[i] > a[j]) {
                    int tmp = a[i];             // swap a[i] and a[j]
                    a[i] = a[j];
                    a[j] = tmp;
                    swaps++;
                    if (trace)
                        row(j == i + 1 ? "" + i : "", "" + j, a, String.format("a[%d]=%d>a[%d]=%d: swap a[%d] <-> a[%d]", i, ai, j, aj, i, j));
                } else if (trace)
                    row(j == i + 1 ? "" + i : "", "" + j, a, String.format("a[%d]=%d<=a[%d]=%d: no swap", i, ai, j, aj));
            }
        }
        return new int[]{comparisons, swaps};
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s %3s  %-20s  %s\n", "i", "j", "a", "action");
        int[] c = exchangeSort(a, n, true);
        System.out.printf("Sorted array: %s\n", Arrays.toString(a));
        System.out.printf("Key comparisons: %d, swaps: %d\n\n", c[0], c[1]);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // “Your turn”: the array comes from the command line
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Your turn", v);
            return;
        }
        run("Instance 1", new int[]{5, 2, 4, 1, 3});
        run("Instance 2", new int[]{12, 11, 13, 5, 6});
    }
}
```
