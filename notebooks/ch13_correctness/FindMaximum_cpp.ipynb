## Find maximum: FindMaximum

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** the maximum of $a$. **Invariant:** every time the condition is evaluated, $max = \max\{a[1],\dots,a[i-1]\}$.

Pseudo-code: section 13.2 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/13-correctness/).

```python
%%writefile find_maximum.cpp
#include <algorithm>
#include <cstdio>
#include <string>
using namespace std;

const int MAXN = 64;

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

bool Equal(const int* x, const int* y, int len)     // x[0..len-1] == y[0..len-1]
{
    for (int k = 0; k < len; k++)
        if (x[k] != y[k]) return false;
    return true;
}

int MaxOf(const int* a, int from, int to)           // maximum of a[from..to-1]
{
    int m = a[from];
    for (int k = from + 1; k < to; k++)
        if (a[k] > m) m = a[k];
    return m;
}

// Invariant: max = maximum of a[1..i-1]
void PrintInvariant(const int* a, int i, int max)
{
    int m = MaxOf(a, 0, i);                         // a[1..i-1] of the pseudo-code
    printf("%2d | %3d | %-26s | %14d | %s\n", i + 1, max, Fmt(a, 0, i).c_str(), m, max == m ? "ok" : "FALSE");
}

// FindMaximum(a, n)
int FindMaximum(int a[], int n)
{
    int max = a[0];
    int i = 1;                                      // i := 2
    PrintInvariant(a, i, max);                      // before the first evaluation of the condition
    while (i < n)                                   // i <= n
    {
        if (a[i] > max)
        {
            max = a[i];
        }
        i = i + 1;
        PrintInvariant(a, i, max);                  // after every iteration, i.e. before the next evaluation
    }
    return max;
}

#ifndef SOLO_LIBRERIA
int main()
{
    int a[] = {9, 12, 3, 16, 19, 11, 12};           // example of the lecture notes
    int n = 7;
    printf("a = %s, n = %d\n", Fmt(a, 0, n).c_str(), n);
    printf(" i | max | a[1..i-1]                  | max{a[1..i-1]} | invariant\n");
    int m = FindMaximum(a, n);
    printf("The loop ends with i = n + 1 = %d: FindMaximum returns %d\n", n + 1, m);
    return 0;
}
#endif
```

**Try it yourself.** Change the array: the invariant holds on every row.

```python
%%writefile prova_find_maximum.cpp
#define SOLO_LIBRERIA   // do not run the main of find_maximum.cpp
#include "find_maximum.cpp"

int main()
{
    int a[] = {4, 8, 8, 1, 10, 2};      // <-- change the array
    int n = 6;
    printf(" i | max | a[1..i-1]                  | max{a[1..i-1]} | invariant\n");
    int m = FindMaximum(a, n);
    printf("returns %d\n", m);
    return 0;
}
```
