## FindMaximum

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** the maximum value in the array.

The variable `max` starts from the first element and is updated whenever a larger element is found.

```python
%%writefile find_maximum.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Returns the maximum of a[0..n-1], with n >= 1
int FindMaximum(int n, int *a, bool trace = false)
{
    int max = a[0];                 // start from the first element
    if (trace)
        printf("%3s  %4s  %4d  %s\n", "--", "--", max, "--");
    for (int i = 1; i < n; i++)     // i = 2..n in the pseudo-code
    {
        bool bigger = a[i] > max;   // comparison, saved for the trace
        if (bigger)
        {
            max = a[i];             // update the maximum
        }
        if (trace)
            printf("%3d  %4d  %4d  %s\n", i, a[i], max, bigger ? "Yes" : "No");
    }
    return max;
}

void Run(string title, vector<int> v)
{
    int n = v.size();
    printf("%s: a = %s, n = %d\n", title.c_str(), Str(n, v.data()).c_str(), n);
    printf("%3s  %4s  %4s  %s\n", "i", "a[i]", "max", "a[i]>max?");
    int result = FindMaximum(n, v.data(), true);
    printf("Return: max = %d\n\n", result);
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // “Your turn”: the array comes from the command line
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        Run("Your turn", v);
        return 0;
    }
    Run("Instance 1", {9, 12, 3, 16, 19, 11, 12});
    Run("Instance 2", {3, 7, 2, 9, 1});
    return 0;
}
```

## CountPositive

**Input:** an array $a[1..n]$ of $n$ integer numbers. **Output:** the number of positive elements in the array.

```python
%%writefile count_positive.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Returns the number of positive elements of a[0..n-1]
int CountPositive(int n, int *a, bool trace = false)
{
    int count = 0;
    if (trace)
        printf("%3s  %4s  %5d  %s\n", "--", "--", count, "--");
    for (int i = 0; i < n; i++)     // i = 1..n in the pseudo-code
    {
        bool positive = a[i] > 0;   // comparison, saved for the trace
        if (positive)
        {
            count = count + 1;
        }
        if (trace)
            printf("%3d  %4d  %5d  %s\n", i, a[i], count, positive ? "Yes" : "No");
    }
    return count;
}

void Run(string title, vector<int> v)
{
    int n = v.size();
    printf("%s: a = %s, n = %d\n", title.c_str(), Str(n, v.data()).c_str(), n);
    printf("%3s  %4s  %5s  %s\n", "i", "a[i]", "count", "a[i]>0?");
    int result = CountPositive(n, v.data(), true);
    printf("Return: count = %d\n\n", result);
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // “Your turn”: the array comes from the command line
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        Run("Your turn", v);
        return 0;
    }
    Run("Instance 1", {-2, 5, -1, 3, 0});
    Run("Instance 2", {4, -3, 7, 2});
    return 0;
}
```

## IsSorted

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** `true` if the values in the array are sorted in non-decreasing order, `false` otherwise.

The `return` statement inside the loop terminates the function as soon as two consecutive elements in the wrong order are found.

```python
%%writefile is_sorted.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Returns true if a[0..n-1] is sorted in non-decreasing order
bool IsSorted(int n, int *a, bool trace = false)
{
    for (int i = 0; i < n - 1; i++)     // i = 1..n-1 in the pseudo-code
    {
        bool wrong = a[i] > a[i + 1];   // comparison, saved for the trace
        if (trace)
            printf("%3d  %4d  %6d  %s\n", i, a[i], a[i + 1], wrong ? "Yes" : "No");
        if (wrong)
        {
            return false;
        }
    }
    return true;
}

void Run(string title, vector<int> v)
{
    int n = v.size();
    printf("%s: a = %s, n = %d\n", title.c_str(), Str(n, v.data()).c_str(), n);
    printf("%3s  %4s  %6s  %s\n", "i", "a[i]", "a[i+1]", "a[i]>a[i+1]?");
    bool result = IsSorted(n, v.data(), true);
    printf("Return: %s\n\n", result ? "true" : "false");
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // “Your turn”: the array comes from the command line
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        Run("Your turn", v);
        return 0;
    }
    Run("Instance 1", {2, 5, 7, 9});
    Run("Instance 2", {3, 1, 4, 2});
    return 0;
}
```

## ScalarProduct

**Input:** two arrays $a[1..n]$ and $b[1..n]$ of size $n \ge 1$. **Output:** the scalar product $\sum_{i=1}^{n} a[i] \cdot b[i]$ of the two arrays.

```python
%%writefile scalar_product.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Returns the scalar product of a[0..n-1] and b[0..n-1], with n >= 1
int ScalarProduct(int n, int *a, int *b, bool trace = false)
{
    int prod = a[0] * b[0];
    if (trace)
        printf("%3s  %4s  %4s  %9s  %5d\n", "--", "--", "--", "--", prod);
    for (int i = 1; i < n; i++)     // i = 2..n in the pseudo-code
    {
        prod = prod + a[i] * b[i];
        if (trace)
            printf("%3d  %4d  %4d  %9d  %5d\n", i, a[i], b[i], a[i] * b[i], prod);
    }
    return prod;
}

void Run(string title, vector<int> a, vector<int> b)
{
    int n = a.size();
    printf("%s: a = %s, b = %s, n = %d\n", title.c_str(), Str(n, a.data()).c_str(), Str(n, b.data()).c_str(), n);
    printf("%3s  %4s  %4s  %9s  %5s\n", "i", "a[i]", "b[i]", "a[i]*b[i]", "prod");
    int result = ScalarProduct(n, a.data(), b.data(), true);
    printf("Return: prod = %d\n\n", result);
}

int main(int argc, char *argv[])
{
    if (argc > 2)   // “Your turn”: the numbers come from the command line (half for a, half for b)
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        int n = v.size() / 2;
        Run("Your turn", vector<int>(v.begin(), v.begin() + n), vector<int>(v.begin() + n, v.begin() + 2 * n));
        return 0;
    }
    Run("Instance 1", {1, 2, 3}, {4, 5, 6});
    Run("Instance 2", {2, 0, -1}, {3, 4, 5});
    return 0;
}
```

## ReverseArray

**Input:** an array $a[1..n]$ of $n$ numbers. **Output:** the same array with the elements in reversed order.

The element $a[i]$ is exchanged with its symmetric element $a[n-i+1]$; only $\lfloor n/2 \rfloor$ exchanges are needed (with $n$ odd, the middle element stays in place).

```python
%%writefile reverse_array.cpp
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Reverses the elements of a[0..n-1] (in place)
void ReverseArray(int n, int *a, bool trace = false)
{
    if (trace)
        printf("%3s  %-14s  %s\n", "--", "--", Str(n, a).c_str());
    for (int i = 0; i < n / 2; i++)     // i = 1..⌊n/2⌋ in the pseudo-code
    {
        int tmp = a[i];
        a[i] = a[n - i - 1];            // a[n-i+1] in the pseudo-code
        a[n - i - 1] = tmp;
        if (trace)
        {
            string ex = "a[" + to_string(i) + "] <-> a[" + to_string(n - i - 1) + "]";
            printf("%3d  %-14s  %s\n", i, ex.c_str(), Str(n, a).c_str());
        }
    }
}

void Run(string title, vector<int> v)
{
    int n = v.size();
    printf("%s: a = %s, n = %d, %d iterations\n", title.c_str(), Str(n, v.data()).c_str(), n, n / 2);
    printf("%3s  %-14s  %s\n", "i", "exchange", "array a (after the exchange)");
    ReverseArray(n, v.data(), true);
    printf("Return: a = %s\n\n", Str(n, v.data()).c_str());
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // “Your turn”: the array comes from the command line
    {
        vector<int> v;
        for (int k = 1; k < argc; k++) v.push_back(atoi(argv[k]));
        Run("Your turn", v);
        return 0;
    }
    Run("Instance 1", {1, 2, 3, 4, 5});
    Run("Instance 2", {10, 20, 30, 40});
    return 0;
}
```
