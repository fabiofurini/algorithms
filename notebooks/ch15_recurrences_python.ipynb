## 1. RecursiveSum and the count of the steps

**RecursiveSum**$(a,n)$ — *Input*: an array $a[1..n]$ of $n\ge1$ integers. *Output*: the sum $a[1]+\dots+a[n]$.

The code follows the pseudo-code (lines 2, 3, 5, 6); a counter `steps` adds up the steps executed by every call, as in the chapter: 2 steps if $n=1$, 3 steps plus the recursive call if $n>1$. The second function is `sumRecDecrease` of the code of the course, which stops at the empty array ($n=0$) and therefore makes one more call.

```python
steps = 0   # steps executed (lines of pseudo-code)
calls = 0   # number of calls

def recursive_sum(a, n):
    """Sum of a[0..n-1] (pseudo-code: a[1..n]), n >= 1."""
    global steps, calls
    calls += 1
    steps += 1                    # line 2: if n = 1
    if n == 1:
        steps += 1                # line 3: return a[1]
        return a[0]
    steps += 2                    # lines 5 and 6
    s = recursive_sum(a, n - 1)
    return s + a[n - 1]

def sum_rec_decrease(a, n):
    """Variant of the course code: base case n = 0."""
    global calls
    calls += 1
    if n == 0:
        return 0
    return sum_rec_decrease(a, n - 1) + a[n - 1]

a = [3, 1, 4, 1, 5, 9, 2, 6]
print("a =", a)
calls = 0
print("RecursiveSum(a, 8) =", recursive_sum(a, len(a)), "with", calls, "calls")
calls = 0
print("sumRecDecrease(a, 8) =", sum_rec_decrease(a, len(a)), "with", calls, "calls")
print("n  measured S(n)  3n-1")
for n in range(1, 9):
    steps = 0
    recursive_sum(a, n)
    print(f"{n:<2d} {steps:>12d} {3 * n - 1:>5d}")
```

## 2. Tower of Hanoi: $T(n)=2T(n-1)+1$

**Hanoi**$(n, from, to, aux)$ — *Input*: the number of disks $n\ge1$ and the three pegs. *Output*: the sequence of moves that transfers disks $1,\dots,n$ from *from* to *to*.

The code counts the moves (and, for $n=3$, prints them): the chapter shows that they are exactly $2^n-1$. The initial call is `Hanoi(n, 'A', 'C', 'B')`.

```python
moves = 0       # number of moves
log = []        # list of the moves (only to print them)

def hanoi(n, src, dst, aux):
    """Hanoi(n, from, to, aux)"""
    global moves
    if n == 1:
        moves += 1                              # move disk 1 from src to dst
        log.append(f"disk 1: {src} -> {dst}")
        return
    hanoi(n - 1, src, aux, dst)
    moves += 1                                  # move disk n
    log.append(f"disk {n}: {src} -> {dst}")
    hanoi(n - 1, aux, dst, src)

hanoi(3, 'A', 'C', 'B')
print("Moves for n = 3:")
for m in log:
    print("  " + m)
print("n   moves  2^n-1")
for n in [1, 2, 3, 4, 5, 6, 7, 8, 10]:
    moves = 0
    log = []
    hanoi(n, 'A', 'C', 'B')
    print(f"{n:<3d} {moves:>5d} {2 ** n - 1:>5d}")
```

## 3. Recurrences computed directly and closed forms

Every recurrence is computed by a recursive function that follows the definition and compared with the closed form obtained in the chapter (expansion method). For the recurrences with $n/2$ the values of $n$ are powers of 2.

| recurrence | closed form |
|---|---|
| $T(n)=T(n-1)+n$, $T(1)=1$ | $n(n+1)/2$ |
| $S(n)=S(n-1)+3$, $S(1)=2$ (RecursiveSum) | $3n-1$ |
| $T(n)=2T(n-1)+1$, $T(1)=1$ (Hanoi) | $2^n-1$ |
| $T(n)=T(n/2)+1$, $T(1)=1$ | $\log_2 n+1$ |
| $T(n)=2T(n/2)+n$, $T(1)=1$ | $n\log_2 n+n$ |

```python
def t_quad(n):  return 1 if n == 1 else t_quad(n - 1) + n
def t_sum(n):   return 2 if n == 1 else t_sum(n - 1) + 3
def t_hanoi(n): return 1 if n == 1 else 2 * t_hanoi(n - 1) + 1
def t_log(n):   return 1 if n == 1 else t_log(n // 2) + 1
def t_nlogn(n): return 1 if n == 1 else 2 * t_nlogn(n // 2) + n

def confronta(nome, valori, ric, chiusa):
    print(nome)
    for n in valori:
        r, c = ric(n), chiusa(n)
        print(f"  n={n:<7d} {r:>12d} {c:>12d}  {'OK' if r == c else 'DIFFERENT'}")

def log2(n):  # n is a power of 2
    return n.bit_length() - 1

confronta("T(n) = T(n-1) + n  ->  n(n+1)/2", [1, 2, 3, 4, 10, 100], t_quad, lambda n: n * (n + 1) // 2)
confronta("S(n) = S(n-1) + 3  ->  3n-1", [1, 2, 3, 4, 100], t_sum, lambda n: 3 * n - 1)
confronta("T(n) = 2T(n-1) + 1  ->  2^n-1", [1, 2, 3, 4, 10, 20], t_hanoi, lambda n: 2 ** n - 1)
confronta("T(n) = T(n/2) + 1  ->  log2(n)+1", [1, 2, 4, 8, 1024, 2 ** 20], t_log, lambda n: log2(n) + 1)
confronta("T(n) = 2T(n/2) + n  ->  n log2(n) + n", [1, 2, 4, 8, 1024, 2 ** 20], t_nlogn, lambda n: n * log2(n) + n)
```

**Try it yourself.** Add a recurrence of your own (here $T(n)=T(n-1)+2n-1$, $T(1)=1$) and compare it with its closed form.

```python
# Add a recurrence: T(n) = T(n-1) + 2n - 1 with T(1) = 1 has closed form n^2.
def t_mia(n): return 1 if n == 1 else t_mia(n - 1) + 2 * n - 1
confronta("T(n) = T(n-1) + 2n - 1  ->  n^2", [1, 2, 3, 10, 100], t_mia, lambda n: n * n)
```
