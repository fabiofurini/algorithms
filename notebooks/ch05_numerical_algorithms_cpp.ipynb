## ChangeOfBase

**Input:** a natural number $n \ge 1$ (in base 10) and a base $b \ge 2$. **Output:** the array $d[1..nd]$ of the digits of $n$ in base $b$ and the number of digits $nd$.

One division by $b$ extracts the least significant digit ($r = n \bmod b$) and leaves as quotient the number written with the other digits: the division is repeated until the quotient is 0. The digits come out from the least significant to the most significant, so in the array `d` the digit of weight $b^0$ is in `d[0]` (in $d[1]$ in the pseudo-code).

The number of digits $nd = \lfloor \log_b n \rfloor + 1$ is computed here with integer multiplications only, without floating point logarithms: in double precision `log(1000)/log(10)` evaluates to $2.9999999999999996$ and the formula would give 3 digits instead of 4.

```python
%%writefile change_of_base.cpp
#include <cmath>
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

// Helper function to print the trace: the first n entries as "[1, 1, 1]"
string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

string Join(const vector<int> &v)       // the entries separated by spaces
{
    string s;
    for (size_t i = 0; i < v.size(); i++) s += (i ? " " : "") + to_string(v[i]);
    return s;
}

// Returns the digits of n in base b: d[0] is the least significant digit, d.size() is nd
vector<int> ChangeOfBase(int n, int b, bool trace = false)
{
    int nd = 1;                         // nd = ⌊log_b n⌋ + 1, with integer operations only
    for (long long p = b; p <= n; p *= b)
        nd = nd + 1;
    vector<int> d(nd);                  // array of the digits
    int k = 0;                          // k = 1 in the pseudo-code
    while (n > 0)
    {
        int q = n / b;                  // quotient
        int r = n % b;                  // remainder (digit in base b)
        d[k] = r;
        n = q;
        k = k + 1;
        if (trace)
            printf("%10d %5d %3d %5d %3d  %s\n", k, q, r, n, k, Str(k, d.data()).c_str());
    }
    return d;
}

string DigitString(const vector<int> &d)    // the digits from the most to the least significant
{
    string s;
    for (int i = d.size() - 1; i >= 0; i--) s += "0123456789ABCDEF"[d[i]];
    return s;
}

void Run(int n, int b)
{
    printf("Conversion: n = %d, b = %d\n", n, b);
    printf("%10s %5s %3s %5s %3s  %s\n", "iteration", "q", "r", "n", "nd", "digits");
    vector<int> d = ChangeOfBase(n, b, true);
    printf("Return: nd = %d, d = %s\n", (int)d.size(), Str(d.size(), d.data()).c_str());
    printf("%d = (%s)_%d\n\n", n, DigitString(d).c_str(), b);
}

int main(int argc, char *argv[])
{
    if (argc > 2)   // “Your turn”: n and the base b from the command line
    {
        Run(atoi(argv[1]), atoi(argv[2]));
        return 0;
    }
    Run(7, 2);
    Run(8, 2);
    Run(25, 3);
    Run(100, 8);
    Run(1000, 10);    // 4 digits: the case where the floating point log(1000)/log(10) fails
    return 0;
}
```

## BasicPrimalityTest

**Input:** a natural number $n$. **Output:** `true` if $n$ is prime, `false` otherwise.

A divisor $d < n$ of $n$ is at most $n/2$: it suffices to test the candidates $d = 2, 3, \dots, \lfloor n/2 \rfloor$. For a prime number the loop is executed completely: for $n = 29$ the candidates are 13. The function also returns the number of candidates tested.

```python
%%writefile basic_primality_test.cpp
#include <cmath>
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

// Helper function to print the trace: the first n entries as "[1, 1, 1]"
string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

string Join(const vector<int> &v)       // the entries separated by spaces
{
    string s;
    for (size_t i = 0; i < v.size(); i++) s += (i ? " " : "") + to_string(v[i]);
    return s;
}

// Returns true if n is prime; tested = number of candidates tested
bool BasicPrimalityTest(int n, int &tested, bool trace = false)
{
    tested = 0;                         // candidates tested
    if (n < 2)
    {
        if (trace)
            printf("%3s %8s  %s\n", "-", "-", "n < 2: return false");
        return false;                   // 0 and 1 are not prime
    }
    for (int d = 2; d <= n / 2; d++)    // d = 2..⌊n/2⌋
    {
        tested++;
        if (n % d == 0)
        {
            if (trace)
                printf("%3d %8d  %s\n", d, n % d, "return false: d divides n");
            return false;               // d divides n: not prime
        }
        if (trace)
            printf("%3d %8d  %s\n", d, n % d, "continue");
    }
    if (trace)
        printf("%3s %8s  %s\n", "-", "-", "the loop ends: return true");
    return true;                        // no divisor found: n is prime
}

void Run(int n)
{
    int tested;
    printf("Check: n = %d, floor(n/2) = %d\n", n, n / 2);
    printf("%3s %8s  %s\n", "d", "n mod d", "action");
    bool result = BasicPrimalityTest(n, tested, true);
    printf("Return: %s, candidates tested: %d\n\n", result ? "true" : "false", tested);
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // “Your turn”: the numbers to check come from the command line
    {
        for (int k = 1; k < argc; k++) Run(atoi(argv[k]));
        return 0;
    }
    Run(36);
    Run(29);
    Run(25);
    return 0;
}
```

## PrimalityTest

**Input:** a natural number $n$. **Output:** `true` if $n$ is prime, `false` otherwise.

The divisors come in pairs $d$ and $n/d$ and in each pair one of the two numbers is at most $\sqrt{n}$: a number $n \ge 2$ is prime if and only if no number in $\{2, 3, \dots, \lfloor \sqrt{n} \rfloor\}$ divides it. For $n = 2$ and $n = 3$ the loop is not executed and the function returns `true`. For $n = 29$ the candidates are 4, instead of the 13 of the basic algorithm.

```python
%%writefile primality_test.cpp
#include <cmath>
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

// Helper function to print the trace: the first n entries as "[1, 1, 1]"
string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

string Join(const vector<int> &v)       // the entries separated by spaces
{
    string s;
    for (size_t i = 0; i < v.size(); i++) s += (i ? " " : "") + to_string(v[i]);
    return s;
}

// Returns true if n is prime; tested = number of candidates tested
bool PrimalityTest(int n, int &tested, bool trace = false)
{
    tested = 0;                         // candidates tested
    if (n < 2)
    {
        if (trace)
            printf("%3s %8s  %s\n", "-", "-", "n < 2: return false");
        return false;                   // 0 and 1 are not prime
    }
    for (int d = 2; d <= (int)sqrt((double)n); d++)     // d = 2..⌊√n⌋
    {
        tested++;
        if (n % d == 0)
        {
            if (trace)
                printf("%3d %8d  %s\n", d, n % d, "return false: d divides n");
            return false;               // d divides n: not prime
        }
        if (trace)
            printf("%3d %8d  %s\n", d, n % d, "continue");
    }
    if (trace)
        printf("%3s %8s  %s\n", "-", "-", "the loop ends: return true");
    return true;                        // no divisor found: n is prime
}

void Run(int n)
{
    int tested;
    printf("Check: n = %d, floor(sqrt(n)) = %d\n", n, (int)sqrt((double)n));
    printf("%3s %8s  %s\n", "d", "n mod d", "action");
    bool result = PrimalityTest(n, tested, true);
    printf("Return: %s, candidates tested: %d\n\n", result ? "true" : "false", tested);
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // “Your turn”: the numbers to check come from the command line
    {
        for (int k = 1; k < argc; k++) Run(atoi(argv[k]));
        return 0;
    }
    Run(36);
    Run(29);
    Run(25);
    Run(2);         // the loop is not executed
    return 0;
}
```

## SieveOfEratosthenes

**Input:** a natural number $n \ge 2$. **Output:** a boolean array $prime[1..n]$ where $prime[i] = 1$ if and only if $i$ is prime.

We start with all the numbers marked as prime (except 1) and, for every prime $p \le \lfloor \sqrt{n} \rfloor$, we mark its multiples as composite starting from $p^2$. In Python, C++ and Java the array starts from 0: `prime[i]` is associated with the number $i$, so it has $n+1$ entries `prime[0..n]` and `prime[0]` (the number 0, not prime) is 0.

```python
%%writefile sieve_of_eratosthenes.cpp
#include <cmath>
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

// Helper function to print the trace: the first n entries as "[1, 1, 1]"
string Str(int n, const int *a)
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

string Join(const vector<int> &v)       // the entries separated by spaces
{
    string s;
    for (size_t i = 0; i < v.size(); i++) s += (i ? " " : "") + to_string(v[i]);
    return s;
}

// Returns prime[0..n]: prime[i] = 1 if and only if i is prime (n >= 2)
vector<int> SieveOfEratosthenes(int n, bool trace = false)
{
    vector<int> prime(n + 1, 1);        // prime[0..n], all set to 1
    prime[0] = 0;                       // 0 is not prime
    prime[1] = 0;                       // 1 is not prime
    if (trace)
        printf("%5s  %s\n", "start", Join(prime).c_str());

    for (int p = 2; p <= (int)sqrt((double)n); p++)     // p = 2..⌊√n⌋
    {
        string js;                      // the values of j (only for the trace)
        if (prime[p] == 1)
        {
            // mark the multiples of p starting from p^2
            for (int j = p * p; j <= n; j += p)
            {
                prime[j] = 0;           // j is not prime
                js += (js.empty() ? "" : ", ") + to_string(j);
            }
        }
        if (trace)
            printf("%5d  %s  %s\n", p, Join(prime).c_str(),
                   js.empty() ? ("no inner loop: prime[" + to_string(p) + "] is 0").c_str() : ("j = " + js).c_str());
    }
    return prime;
}

void Run(int n)
{
    printf("Sieve: n = %d, floor(sqrt(n)) = %d\n", n, (int)sqrt((double)n));
    printf("%5s  prime[0..%d]\n", "p", n);
    vector<int> prime = SieveOfEratosthenes(n, true);
    string primes = "[";
    for (int i = 0; i <= n; i++)
        if (prime[i] == 1) primes += (primes.size() > 1 ? ", " : "") + to_string(i);
    printf("Prime numbers <= %d: %s]\n\n", n, primes.c_str());
}

int main(int argc, char *argv[])
{
    if (argc > 1)   // “Your turn”: n from the command line
    {
        Run(atoi(argv[1]));
        return 0;
    }
    Run(10);
    Run(20);
    return 0;
}
```

## EuclideanAlgorithm

**Input:** two natural numbers $n, m \ge 1$. **Output:** $\gcd(n, m)$, the greatest common divisor.

The pair $(n, m)$ is replaced with $(m,\ n \bmod m)$, which has the same common divisors, until the second number is 0: then $\gcd(n, 0) = n$. The parameters are passed by value: the function modifies its own copies. Two numbers are coprime if and only if the result is 1.

```python
%%writefile euclidean_algorithm.cpp
#include <cstdio>
#include <cstdlib>

// Returns the greatest common divisor of n and m (n, m >= 1)
int EuclideanAlgorithm(int n, int m, bool trace = false)
{
    int step = 0;                       // only for the trace
    while (m != 0)
    {
        step++;
        int r = n % m;
        if (trace)
            printf("%5d %5d %5d  %d mod %d = %d\n", step, n, m, n, m, r);
        n = m;
        m = r;
    }
    if (trace)
        printf("%5d %5d %5d  %s\n", step + 1, n, m, "--- (the loop ends)");
    return n;                           // m = 0: n is the GCD
}

void Run(int n, int m)
{
    printf("GCD computation: n = %d, m = %d\n", n, m);
    printf("%5s %5s %5s  %s\n", "step", "n", "m", "r = n mod m");
    int g = EuclideanAlgorithm(n, m, true);     // n and m are unchanged: passed by value
    printf("Return: gcd(%d, %d) = %d\n", n, m, g);
    printf("coprime: %s\n\n", g == 1 ? "true" : "false");
}

int main(int argc, char *argv[])
{
    if (argc > 2)   // “Your turn”: the two numbers from the command line
    {
        Run(atoi(argv[1]), atoi(argv[2]));
        return 0;
    }
    Run(48, 18);
    Run(252, 105);
    Run(35, 12);
    return 0;
}
```
