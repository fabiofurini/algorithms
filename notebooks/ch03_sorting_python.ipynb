## InsertionSort

**Input:** an array $a[1..n]$ of $n$ natural numbers. **Output:** the array $a$ sorted in non-decreasing order.

At step $i$ the sub-array $a[1..i-1]$ is sorted: the key $a[i]$ is inserted in its place, after shifting one position to the right the greater keys. The function also returns the counters used in the lecture notes: the number of key comparisons and the number of shifts (no swaps: a shift is a single assignment).

```python
def row(i, j, key, a, action):          # one row of the trace
    print("%3s %3s %4s  %-20s  %s" % (i, j, key, a, action))


def insertion_sort(a, n, trace=False):
    comparisons = 0                     # key comparisons
    shifts = 0                          # shifts
    if trace:
        row("-", "-", "-", a, "initial state")
    i = 1                               # i = 2 in the pseudo-code
    while i < n:                        # i <= n in the pseudo-code
        key = a[i]                      # element to insert
        j = i - 1
        if trace:
            row(i, j, key, a, "key <- a[%d]=%d; j <- %d" % (i, key, j))
        while j >= 0 and a[j] > key:    # j >= 1 in the pseudo-code
            comparisons += 1
            shifts += 1
            a[j + 1] = a[j]             # shift the element to the right
            j = j - 1
            if trace:
                row("", j, key, a, "a[%d]=%d>%d: shift a[%d] <- %d; j <- %d" % (j + 1, a[j + 1], key, j + 2, a[j + 1], j))
        if j >= 0:
            comparisons += 1            # the comparison that stopped the loop (a[j] <= key)
        a[j + 1] = key                  # insert key in its position
        if trace:
            if j < 0:
                row("", "-", key, a, "j=%d: exit; insert a[%d] <- %d" % (j, j + 1, key))
            else:
                row("", "-", key, a, "a[%d]=%d<=%d: exit; insert a[%d] <- %d" % (j, a[j], key, j + 1, key))
        i = i + 1
    return comparisons, shifts
```

### Execution on the examples of the chapter

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s %3s %4s  %-20s  %s" % ("i", "j", "key", "a", "action"))
    comparisons, shifts = insertion_sort(a, n, trace=True)
    print("Sorted array: %s" % a)
    print("Key comparisons: %d, shifts: %d" % (comparisons, shifts))
    print()


run("Instance 1", [5, 2, 4, 1, 3])
run("Instance 2", [12, 11, 13, 5, 6])
```

## ExchangeSort

**Input:** an array $a[1..n]$ of $n$ natural numbers. **Output:** the array $a$ sorted in non-decreasing order.

At step $i$ the element $a[i]$ is compared with every $a[j]$ with $j > i$, and the two elements are swapped whenever $a[i] > a[j]$: at the end of the step $a[i]$ is the minimum of $a[i..n]$. The function returns the number of key comparisons and the number of swaps.

```python
def row(i, j, a, action):                # one row of the trace
    print("%3s %3s  %-20s  %s" % (i, j, a, action))


def exchange_sort(a, n, trace=False):
    comparisons = 0                     # key comparisons
    swaps = 0                           # swaps
    if trace:
        row("-", "-", a, "initial state")
    for i in range(n - 1):              # i = 1..n-1 in the pseudo-code
        for j in range(i + 1, n):       # j = i+1..n in the pseudo-code
            comparisons += 1
            ai, aj = a[i], a[j]         # values before the swap, for the trace
            if a[i] > a[j]:
                tmp = a[i]              # swap a[i] and a[j]
                a[i] = a[j]
                a[j] = tmp
                swaps += 1
                if trace:
                    row(i if j == i + 1 else "", j, a, "a[%d]=%d>a[%d]=%d: swap a[%d] <-> a[%d]" % (i, ai, j, aj, i, j))
            elif trace:
                row(i if j == i + 1 else "", j, a, "a[%d]=%d<=a[%d]=%d: no swap" % (i, ai, j, aj))
    return comparisons, swaps
```

## SelectionSort

**Input:** an array $a[1..n]$ of $n$ natural numbers. **Output:** the array $a$ sorted in non-decreasing order.

At step $i$ the inner loop looks for the index $\mathit{minIdx}$ of the minimum of $a[i..n]$ (the comparison is strict: in case of ties the first minimum wins) and then, if $\mathit{minIdx} \ne i$, swaps $a[i]$ with $a[\mathit{minIdx}]$: at most one swap per step. The function returns the number of key comparisons and the number of swaps.

```python
def row(i, j, min_idx, a, action):      # one row of the trace
    print("%3s %3s %6s  %-20s  %s" % (i, j, min_idx, a, action))


def selection_sort(a, n, trace=False):
    comparisons = 0                     # key comparisons
    swaps = 0                           # swaps
    if trace:
        row("-", "-", "-", a, "initial state")
    for i in range(n - 1):              # i = 1..n-1 in the pseudo-code
        min_idx = i                     # index of the minimum in a[i..n]
        if trace:
            row(i, "-", min_idx, a, "minIdx := %d (a[%d]=%d)" % (i, i, a[i]))
        for j in range(i + 1, n):       # j = i+1..n in the pseudo-code
            comparisons += 1
            if a[j] < a[min_idx]:
                old = min_idx           # only for the trace
                min_idx = j             # new minimum found
                if trace:
                    row("", j, min_idx, a, "a[%d]=%d<a[%d]=%d: minIdx <- %d" % (j, a[j], old, a[old], j))
            elif trace:
                row("", j, min_idx, a, "a[%d]=%d>=a[%d]=%d: no update" % (j, a[j], min_idx, a[min_idx]))
        if min_idx != i:
            tmp = a[i]                  # swap a[i] and a[minIdx]
            a[i] = a[min_idx]
            a[min_idx] = tmp
            swaps += 1
            if trace:
                row("", "-", "-", a, "minIdx=%d != %d: swap a[%d] <-> a[%d]" % (min_idx, i, i, min_idx))
        elif trace:
            row("", "-", "-", a, "minIdx=%d: no swap" % min_idx)
    return comparisons, swaps
```

### Execution on the examples of the chapter

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s %3s %6s  %-20s  %s" % ("i", "j", "minIdx", "a", "action"))
    comparisons, swaps = selection_sort(a, n, trace=True)
    print("Sorted array: %s" % a)
    print("Key comparisons: %d, swaps: %d" % (comparisons, swaps))
    print()


run("Instance 1", [5, 2, 4, 1, 3])
run("Instance 2", [12, 11, 13, 5, 6])
```

### Comparison of the three algorithms

On the same two arrays, the three algorithms perform (as in the comparison example of the lecture notes):

| algorithm | $[5, 2, 4, 1, 3]$: comparisons | moves | $[12, 11, 13, 5, 6]$: comparisons | moves |
|---|---|---|---|---|
| Insertion Sort | 9 | 7 shifts | 9 | 7 shifts |
| Exchange Sort | 10 | 7 swaps | 10 | 7 swaps |
| Selection Sort | 10 | 3 swaps | 10 | 3 swaps |

Selection Sort always performs $n(n-1)/2$ comparisons but at most $n-1$ swaps; Insertion Sort adapts to the input (on an already sorted array it performs only $n-1$ comparisons and no shift) and is the only stable one of the three.
