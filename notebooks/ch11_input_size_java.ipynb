## 1. The input size of the examples

**Input:** a natural number, two natural numbers or an array of natural numbers. **Output:** the input size $|I|$, i.e. the number of bits needed to write it in binary ($\lfloor \log_2 n \rfloor + 1$ bits for a number $n \ge 1$).

See: sections 11.1, 11.3, 11.4, 11.5 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/11-input-size/).

```python
%%writefile InputSize.java
public class InputSize
{
static long[] cnt = new long[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Line | Executions");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Total steps S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    // Number of bits of a natural n >= 1: floor(log2 n) + 1
    static int bits(int n)
    {
        int k = 0;
        while (n > 0)
        {
            n = n / 2;        // halve until reaching 0
            k = k + 1;
        }
        return k;
    }

    // Digits of n in base 2 (as a string)
    static String binary(int n)
    {
        String s = "";
        while (n > 0)
        {
            s = (n % 2) + s;
            n = n / 2;
        }
        return s;
    }

    static double log2(double x)
    {
        return Math.log(x) / Math.log(2);
    }

    public static void main(String[] args)
    {
        // Example of the lecture notes: the number of bits of some instances
        int n = 37, m = 200;
        System.out.printf("One number: n = %d = (%s)_2, |I| = %d\n", n, binary(n), bits(n));
        System.out.printf("Two numbers: n = %d, m = %d = (%s)_2, |I| = %d + %d = %d\n", n, m, binary(m), bits(n), bits(m), bits(n) + bits(m));
        int[] a = {6, 1, 3, 8};
        int[] b = new int[4];
        int sum = 0, M = 0;
        for (int i = 0; i < 4; i++)
        {
            b[i] = bits(a[i]);
            sum += b[i];
            if (a[i] > M) M = a[i];
        }
        System.out.printf("One array: a = %s, bits per entry = %s, |I| = %d\n", fmt(a, 0, 4), fmt(b, 0, 4), sum);

        // The bounds of the lecture notes (values rounded to two decimals)
        System.out.printf("\n");
        System.out.printf("One number: log2(n) = %.2f <= |I| = %d <= 2 log2(n) = %.2f\n", log2(n), bits(n), 2 * log2(n));
        double s = log2(n) + log2(m);
        System.out.printf("Two numbers: log2(n) + log2(m) = %.2f <= |I| = %d <= 2 (log2(n) + log2(m)) = %.2f\n", s, bits(n) + bits(m), 2 * s);
        int mx = Math.max(n, m);
        System.out.printf("Two numbers and maximum: log2(max) = %.2f <= |I| = %d <= 4 log2(max) = %.2f\n", log2(mx), bits(n) + bits(m), 4 * log2(mx));
        System.out.printf("One array: n = %d <= |I| = %d <= n (1 + log2 M) = %.2f\n", 4, sum, 4 * (1 + log2(M)));

        // The upper bound is not tight: a[1] = M = 2^n, a[2..n] = 1 has |I| = 2n, but n(1 + log2 M) = n(n + 1)
        System.out.printf("\n");
        System.out.printf("   n |      M |  |I| = 2n | n (1 + log2 M)\n");
        for (int k = 1; k <= 10; k++)
        {
            int big = 1 << k;
            int size = bits(big) + (k - 1);     // a[1] = M, a[2..n] = 1 (1 bit each)
            System.out.printf("%4d | %6d | %9d | %13d\n", k, big, size, k * (k + 1));
        }
    }
}
```

## 2. Sum of the entries of an array: SumArray

**Input:** an array $a[1..n]$ of $n \ge 1$ natural numbers. **Output:** $a[1]+a[2]+\cdots+a[n]$. Number of steps: $S(n) = 3n + 4$, whatever the values.

Pseudo-code: section 11.6 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/11-input-size/).

```python
%%writefile SumArray.java
public class SumArray
{
static long[] cnt = new long[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Line | Executions");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Total steps S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    // SumArray(a, n)
    static int sumArray(int[] a)
    {
        int n = a.length;
        int s = 0; cnt[2]++;
        int i = 0; cnt[3]++;               // i := 1 in the pseudo-code
        cnt[4]++;                          // the while condition is counted at every evaluation
        while (i < n)                      // i <= n in the pseudo-code
        {
            s = s + a[i]; cnt[5]++;
            i = i + 1; cnt[6]++;
            cnt[4]++;
        }
        cnt[8]++;
        return s;
    }

    public static void main(String[] args)
    {
        int[] a = {6, 1, 3, 8};
        int n = a.length;
        reset();
        int s = sumArray(a);
        System.out.printf("a = %s, n = %d, sum = %d\n", fmt(a, 0, n), n, s);
        printSteps(2, 3, 4, 5, 6, 8);
        System.out.printf("Formula: S(n) = 3n + 4 = %d\n", 3 * n + 4);

        // Worst case: the array of |I| ones has input size |I| and takes exactly 3|I| + 4 steps
        System.out.printf("\n");
        System.out.printf("Array of |I| ones:\n");
        int[] sizes = {4, 8, 16, 32};
        for (int size : sizes)
        {
            int[] ones = new int[size];
            java.util.Arrays.fill(ones, 1);
            reset();
            sumArray(ones);
            System.out.printf("|I| = %2d: S = %3d = 3|I| + 4 = %d\n", size, total(), 3 * size + 4);
        }
    }
}
```

```python
%%writefile ProvaSumArray.java
public class ProvaSumArray
{
    public static void main(String[] args)
    {
        int[] a = {5, 1, 9, 2, 7};      // <-- change the array
        SumArray.reset();
        int s = SumArray.sumArray(a);
        System.out.printf("sum = %d, S = %d, 3n + 4 = %d\n", s, SumArray.total(), 3 * a.length + 4);
    }
}
```
