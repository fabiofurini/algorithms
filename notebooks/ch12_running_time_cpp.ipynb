The running time is measured in **steps**: a step is the execution of one line of pseudo-code, and the condition of a `while` counts as one step each time it is evaluated. The lines are numbered as in the lecture notes (line 1 is the `function` header) and an array `cnt` counts how many times each line is executed: the printed tables are those of the chapter.

Every program is a `%%writefile` cell that writes the source file, followed by a cell that compiles it with `g++` (C++17) and runs it.

**Indices.** In the pseudo-code the arrays are indexed from 1; in the C++ code from 0: the element `a[i]` of the pseudo-code is `a[i-1]` in the code (and a loop `while i <= n` from 1 becomes `while i < n` from 0). The printed tables use the indices of the pseudo-code.

## 1. Find maximum: FindMaximum

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** the maximum value of $a$. In the worst case (increasing array) $S(n) = 4n = \Theta(n)$.

Pseudo-code: section 12.2 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/12-running-time/).

```python
%%writefile find_maximum.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Line | Executions\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Total steps S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// FindMaximum(a, n)
int FindMaximum(int a[], int n)
{
    int max = a[0]; cnt[2]++;          // max := a[1]
    int i = 1; cnt[3]++;               // i := 2 in the pseudo-code
    cnt[4]++;                          // the while condition is counted at every evaluation
    while (i < n)                      // i <= n in the pseudo-code
    {
        cnt[5]++;
        if (a[i] > max)
        {
            max = a[i]; cnt[6]++;
        }
        i = i + 1; cnt[8]++;
        cnt[4]++;
    }
    cnt[10]++;
    return max;
}

long long StepsOf(int a[], int n)
{
    Reset();
    FindMaximum(a, n);
    return Total();
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Example of the lecture notes: a[1..4] = (3, 1, 4, 5)
    int a[] = {3, 1, 4, 5};
    int n = 4;
    Reset();
    int m = FindMaximum(a, n);
    printf("a = %s -> max = %d\n", Fmt(a, 0, n).c_str(), m);
    PrintSteps({2, 3, 4, 5, 6, 8, 10});
    long long t = cnt[6];
    printf("sum t(i) = %lld, S(n) = 3n + 1 + sum t(i) = %lld\n", t, 3 * n + 1 + t);

    // Worst case (increasing array) and best case (decreasing array)
    printf("\n");
    int inc[] = {1, 2, 3, 4};
    printf("a = %s (increasing): S = %lld = 4n = %d\n", Fmt(inc, 0, n).c_str(), StepsOf(inc, n), 4 * n);
    int dec[] = {4, 3, 2, 1};
    printf("a = %s (decreasing): S = %lld = 3n + 1 = %d\n", Fmt(dec, 0, n).c_str(), StepsOf(dec, n), 3 * n + 1);

    printf("\n");
    printf("Increasing array (1, 2, ..., n):\n");
    printf("  n | S(n) | 4n\n");
    for (int k = 1; k <= 8; k++)
    {
        int b[8];
        for (int i = 0; i < k; i++) b[i] = i + 1;
        printf("%3d | %4lld | %d\n", k, StepsOf(b, k), 4 * k);
    }
    return 0;
}
#endif
```

## 2. Insertion sort: InsertionSort

**Input:** an array $a[1..n]$ of $n$ natural numbers. **Output:** the array $a$ sorted in non-decreasing order. If $t(i)$ is the number of shifts at iteration $i$, then $S(n) = 6n - 4 + 3\sum_{i=2}^{n} t(i)$; in the worst case (decreasing array) $S(n) = \frac{3n^2+9n-8}{2} = \Theta(n^2)$.

Pseudo-code: section 12.3 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/12-running-time/).

```python
%%writefile insertion_sort.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Line | Executions\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Total steps S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// InsertionSort(a, n)
void InsertionSort(int a[], int n, bool trace = false)
{
    long long prev = 0;                    // only for the trace
    int i = 1; cnt[2]++;                   // i := 2 in the pseudo-code
    cnt[3]++;
    while (i < n)                          // i <= n in the pseudo-code
    {
        int key = a[i]; cnt[4]++;
        int j = i - 1; cnt[5]++;
        cnt[6]++;                          // the while condition is counted at every evaluation
        while (j >= 0 && a[j] > key)       // j >= 1 and a[j] > key in the pseudo-code
        {
            a[j + 1] = a[j]; cnt[7]++;
            j = j - 1; cnt[8]++;
            cnt[6]++;
        }
        a[j + 1] = key; cnt[10]++;
        i = i + 1; cnt[11]++;
        cnt[3]++;
        if (trace)                         // trace: after the increment, i is the pseudo-code index of the iteration just finished
        {
            printf("%2d | %3d | %4lld | %s\n", i, key, cnt[7] - prev, Fmt(a, 0, n).c_str());
            prev = cnt[7];
        }
    }
}

long long StepsOf(const int a[], int n)    // works on a copy of a
{
    int b[64];
    for (int i = 0; i < n; i++) b[i] = a[i];
    Reset();
    InsertionSort(b, n);
    return Total();
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Example of the lecture notes: a[1..6] = (5, 2, 4, 6, 1, 3)
    int a[] = {5, 2, 4, 6, 1, 3};
    int n = 6;
    printf("a = %s\n", Fmt(a, 0, n).c_str());
    Reset();
    printf(" i | key | t(i) | array after the iteration\n");
    InsertionSort(a, n, true);
    printf("sorted: %s\n", Fmt(a, 0, n).c_str());
    PrintSteps({2, 3, 4, 5, 6, 7, 8, 10, 11});
    long long t = cnt[7];
    printf("sum t(i) = %lld, S(n) = 6n - 4 + 3 sum t(i) = %lld\n", t, 6 * n - 4 + 3 * t);

    // Worst case: decreasing array, t(i) = i - 1
    printf("\n");
    int d[] = {6, 5, 4, 3, 2, 1};
    printf("a = %s (decreasing): S = %lld = (3n^2 + 9n - 8)/2 = %d\n", Fmt(d, 0, n).c_str(), StepsOf(d, n), (3 * n * n + 9 * n - 8) / 2);
    printf("sum t(i) = %lld = n(n-1)/2 = %d\n", cnt[7], n * (n - 1) / 2);

    printf("\n");
    printf("Decreasing array (n, ..., 2, 1):\n");
    printf("  n | S(n) | (3n^2 + 9n - 8)/2\n");
    for (int k = 1; k <= 8; k++)
    {
        int b[8];
        for (int i = 0; i < k; i++) b[i] = k - i;
        printf("%3d | %4lld | %d\n", k, StepsOf(b, k), (3 * k * k + 9 * k - 8) / 2);
    }
    return 0;
}
#endif
```

## 3. Change of base: ChangeToBase2

**Input:** a natural number $n \ge 1$ in base 10. **Output:** the array $d[0..nd-1]$ of binary digits and the number of digits $nd$. For every $n$: $S(n) = 6\lfloor\log_2 n\rfloor + 11 = \Theta(\log_2 n)$.

Pseudo-code: section 12.4 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/12-running-time/).

```python
%%writefile change_to_base2.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Line | Executions\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Total steps S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

int FloorLog2(int n)                       // floor(log2 n) computed with a loop (it does not count as a step)
{
    int k = 0;
    while (n > 1)
    {
        n = n / 2;
        k = k + 1;
    }
    return k;
}

// ChangeToBase2(n): returns the array d[0..nd-1] (to be freed with delete[]) and nd
int* ChangeToBase2(int n, int& nd, bool trace = false)
{
    nd = FloorLog2(n) + 1; cnt[2]++;
    int* d = new int[nd]; cnt[3]++;        // array d[0..nd-1], indexed from 0 as in the pseudo-code
    int k = 0; cnt[4]++;
    cnt[5]++;                              // the while condition is counted at every evaluation
    while (n > 0)
    {
        int q = n / 2; cnt[6]++;
        int r = n % 2; cnt[7]++;
        d[k] = r; cnt[8]++;
        if (trace)                         // trace
        {
            printf("%10d | %3d | %3d | %3d | d[%d] = %d\n", k + 1, n, q, r, k, r);
        }
        n = q; cnt[9]++;
        k = k + 1; cnt[10]++;
        cnt[5]++;
    }
    cnt[12]++;
    return d;
}

string Digits(const int* d, int nd)        // d[nd-1..0] (from the most significant digit)
{
    string s = "";
    for (int k = nd - 1; k >= 0; k--) s += to_string(d[k]);
    return s;
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Example of the lecture notes: n = 13
    int n = 13, nd;
    Reset();
    printf(" iteration |   n |   q |   r | assignment\n");
    int* d = ChangeToBase2(n, nd, true);
    printf("n = %d = (%s)_2: nd = %d, digits d[nd-1..0] = %s\n", n, Digits(d, nd).c_str(), nd, Digits(d, nd).c_str());
    delete[] d;
    PrintSteps({2, 3, 4, 5, 6, 7, 8, 9, 10, 12});
    printf("S(n) = 6 floor(log2 n) + 11 = %d\n", 6 * FloorLog2(n) + 11);

    // The number of steps grows as log2 n
    printf("\n");
    printf("       n | nd | S(n) | 6 floor(log2 n) + 11\n");
    int values[] = {1, 2, 13, 100, 1000, 1000000};
    for (int v : values)
    {
        Reset();
        d = ChangeToBase2(v, nd);
        printf("%8d | %2d | %4lld | %d\n", v, nd, Total(), 6 * FloorLog2(v) + 11);
        delete[] d;
    }
    return 0;
}
#endif
```

## 4. Primality test: PrimalityTest

**Input:** a natural number $n$. **Output:** `true` if $n$ is prime, `false` otherwise. In the worst case ($n$ prime) $S(n) = 3\lfloor\sqrt{n}\rfloor + 1 = \Theta(\sqrt{n})$, i.e. $2^{\Theta(|I|)}$ as a function of the input size.

Pseudo-code: section 12.5 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/12-running-time/).

```python
%%writefile primality_test.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Line | Executions\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Total steps S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

long long FloorSqrt(long long n)           // floor(sqrt(n)) in integer arithmetic
{
    long long r = (long long) sqrtl((long double) n);
    while (r * r > n) r--;
    while ((r + 1) * (r + 1) <= n) r++;
    return r;
}

// PrimalityTest(n)
bool PrimalityTest(long long n)
{
    cnt[2]++;
    if (n < 2)
    {
        cnt[3]++;
        return false;
    }
    long long d = 2; cnt[5]++;
    cnt[6]++;                              // the while condition is counted at every evaluation
    while (d <= FloorSqrt(n))
    {
        cnt[7]++;
        if (n % d == 0)
        {
            cnt[8]++;
            return false;
        }
        d = d + 1; cnt[10]++;
        cnt[6]++;
    }
    cnt[12]++;
    return true;
}

int Bits(long long n)                      // number of bits of n >= 1
{
    int k = 0;
    while (n > 0)
    {
        n = n / 2;
        k = k + 1;
    }
    return k;
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Example of the lecture notes: a prime and a composite number
    long long n = 97;
    Reset();
    bool r = PrimalityTest(n);
    printf("n = %lld: %s\n", n, r ? "true" : "false");
    PrintSteps({2, 3, 5, 6, 7, 8, 10, 12});
    printf("S(n) = 3 floor(sqrt(n)) + 1 = %lld\n", 3 * FloorSqrt(n) + 1);

    printf("\n");
    n = 91;
    Reset();
    r = PrimalityTest(n);
    long long p = 2;
    while (n % p != 0) p = p + 1;
    printf("n = %lld: %s, p = %lld\n", n, r ? "true" : "false", p);
    PrintSteps({2, 3, 5, 6, 7, 8, 10, 12});
    printf("S(n) = 3p - 1 = %lld, while floor(sqrt(n)) = %lld\n", 3 * p - 1, FloorSqrt(n));

    printf("\n");
    n = 1;
    Reset();
    r = PrimalityTest(n);
    printf("n = %lld: %s, S = %lld\n", n, r ? "true" : "false", Total());

    // The worst case are prime numbers: S grows as sqrt(n), i.e. it doubles for every 2 more bits
    printf("\n");
    printf("|I| |        n |  S(n) | 3 floor(sqrt(n)) + 1\n");
    long long primes[] = {1021, 4093, 16381, 65521, 262139, 1048573, 4194301, 16777213};
    for (long long q : primes)
    {
        Reset();
        PrimalityTest(q);
        printf("%3d | %8lld | %5lld | %lld\n", Bits(q), q, Total(), 3 * FloorSqrt(q) + 1);
    }
    return 0;
}
#endif
```
