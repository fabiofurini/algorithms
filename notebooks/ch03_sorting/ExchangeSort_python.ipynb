## ExchangeSort

**Input:** an array $a[1..n]$ of $n$ natural numbers. **Output:** the array $a$ sorted in non-decreasing order.

At step $i$ the element $a[i]$ is compared with every $a[j]$ with $j > i$, and the two elements are swapped whenever $a[i] > a[j]$: at the end of the step $a[i]$ is the minimum of $a[i..n]$. The function returns the number of key comparisons and the number of swaps.

```python
def row(i, j, a, action):                # one row of the trace
    print("%3s %3s  %-20s  %s" % (i, j, a, action))


def exchange_sort(a, n, trace=False):
    comparisons = 0                     # key comparisons
    swaps = 0                           # swaps
    if trace:
        row("-", "-", a, "initial state")
    for i in range(n - 1):              # i = 1..n-1 in the pseudo-code
        for j in range(i + 1, n):       # j = i+1..n in the pseudo-code
            comparisons += 1
            ai, aj = a[i], a[j]         # values before the swap, for the trace
            if a[i] > a[j]:
                tmp = a[i]              # swap a[i] and a[j]
                a[i] = a[j]
                a[j] = tmp
                swaps += 1
                if trace:
                    row(i if j == i + 1 else "", j, a, "a[%d]=%d>a[%d]=%d: swap a[%d] <-> a[%d]" % (i, ai, j, aj, i, j))
            elif trace:
                row(i if j == i + 1 else "", j, a, "a[%d]=%d<=a[%d]=%d: no swap" % (i, ai, j, aj))
    return comparisons, swaps
```

### Execution on the examples of the chapter

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s %3s  %-20s  %s" % ("i", "j", "a", "action"))
    comparisons, swaps = exchange_sort(a, n, trace=True)
    print("Sorted array: %s" % a)
    print("Key comparisons: %d, swaps: %d" % (comparisons, swaps))
    print()


run("Instance 1", [5, 2, 4, 1, 3])
run("Instance 2", [12, 11, 13, 5, 6])
```
