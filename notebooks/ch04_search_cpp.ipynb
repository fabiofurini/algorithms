## LinearSearch

**Input:** an array $a[1..n]$ of $n$ natural numbers and a target value $m$. **Output:** the index of $m$ in $a$, or $-1$ if $m$ is not found.

The algorithm compares $m$ with the elements of the array one after the other and stops at the first one equal to $m$: the returned index is the smallest $k$ with $a[k]=m$. In the code the indices start from 0, so the returned index is the one of the pseudo-code minus 1.

```python
%%writefile linear_search.cpp
#include <cstdarg>
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

// Helper functions to print the trace
string Str(int n, const int *a)     // the array as "[4, 2, 7]"
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Returns the index of m in a[0..n-1], or -1 if m is not found
int LinearSearch(int n, int *a, int m, bool trace = false)
{
    for (int i = 0; i < n; i++)         // i = 1..n in the pseudo-code
    {
        if (a[i] == m)
        {
            if (trace)
                printf("%10d %3d  %-16s  %s\n", i + 1, i, ("a[" + to_string(i) + "] = " + to_string(a[i]) + " = m").c_str(), ("Found! Return " + to_string(i)).c_str());
            return i;                   // element found at index i
        }
        if (trace)
            printf("%10d %3d  %-16s  %s\n", i + 1, i, ("a[" + to_string(i) + "] = " + to_string(a[i]) + " != m").c_str(), "continue");
    }
    if (trace)
        printf("%10s %3d  %-16s  %s\n", "-", n, "-", "i = n: the loop ends, not found! Return -1");
    return -1;                          // element not found
}

void Run(vector<int> v, int m)
{
    int n = v.size();
    printf("Search: m = %d, a = %s, n = %d\n", m, Str(n, v.data()).c_str(), n);
    printf("%10s %3s  %-16s  %s\n", "iteration", "i", "comparison", "action");
    int result = LinearSearch(n, v.data(), m, true);
    printf("Return: %d\n\n", result);
}

int main(int argc, char *argv[])
{
    if (argc > 2)   // “Your turn”: first number = m, the others = array
    {
        vector<int> v;
        for (int k = 2; k < argc; k++) v.push_back(atoi(argv[k]));
        Run(v, atoi(argv[1]));
        return 0;
    }
    vector<int> a = {4, 2, 7, 1, 5, 9, 3};
    Run(a, 5);
    Run(a, 6);
    Run({1, 2, 2, 2, 2, 2, 3}, 2);      // m appears several times: returns the first index
    return 0;
}
```

## BinarySearch

**Input:** an array $a[1..n]$ of $n$ natural numbers **sorted in non-decreasing order** and a target value $m$. **Output:** the index of $m$ in $a$, or $-1$ if $m$ is not found.

The algorithm keeps the search interval $a[left..right]$ and at each iteration compares $m$ with the middle element $a[mid]$, halving the interval: it executes at most $\lfloor \log_2 n \rfloor + 1$ iterations. If $m$ appears more than once it returns the index of one of its occurrences, not necessarily the first one.

```python
%%writefile binary_search.cpp
#include <cstdarg>
#include <cstdio>
#include <cstdlib>
#include <string>
#include <vector>
using namespace std;

// Helper functions to print the trace
string Str(int n, const int *a)     // the array as "[4, 2, 7]"
{
    string s = "[";
    for (int i = 0; i < n; i++) s += (i ? ", " : "") + to_string(a[i]);
    return s + "]";
}

// Returns the index of m in a[0..n-1] (sorted), or -1 if m is not found
int BinarySearch(int n, int *a, int m, bool trace = false)
{
    int left = 0;                       // left boundary (1 in the pseudo-code)
    int right = n - 1;                  // right boundary (n in the pseudo-code)
    int iteration = 0;                  // only for the trace

    while (left <= right)
    {
        iteration++;
        int mid = left + (right - left) / 2;    // middle index

        if (a[mid] == m)
        {
            if (trace)
                printf("%10d %5d %5d %4d  a[%d] = %d = m => Found! Return %d\n", iteration, left, right, mid, mid, a[mid], mid);
            return mid;                 // element found
        }

        if (a[mid] < m)
        {
            if (trace)
                printf("%10d %5d %5d %4d  a[%d] = %d < %d => left = %d + 1 = %d\n", iteration, left, right, mid, mid, a[mid], m, mid, mid + 1);
            left = mid + 1;             // search in the right half
        }
        else
        {
            if (trace)
                printf("%10d %5d %5d %4d  a[%d] = %d > %d => right = %d - 1 = %d\n", iteration, left, right, mid, mid, a[mid], m, mid, mid - 1);
            right = mid - 1;            // search in the left half
        }
    }
    if (trace)
        printf("%10s %5d %5d %4s  left > right => Not found! Return -1\n", "-", left, right, "-");
    return -1;                          // element not found
}

void Run(vector<int> v, int m)
{
    int n = v.size();
    printf("Search: m = %d, a = %s, n = %d\n", m, Str(n, v.data()).c_str(), n);
    printf("%10s %5s %5s %4s  %s\n", "iteration", "left", "right", "mid", "action");
    int result = BinarySearch(n, v.data(), m, true);
    printf("Return: %d\n\n", result);
}

int main(int argc, char *argv[])
{
    if (argc > 2)   // “Your turn”: first number = m, the others = sorted array
    {
        vector<int> v;
        for (int k = 2; k < argc; k++) v.push_back(atoi(argv[k]));
        Run(v, atoi(argv[1]));
        return 0;
    }
    vector<int> a = {3, 9, 11, 12, 12, 16, 19};
    Run(a, 12);
    Run(a, 10);
    Run(a, 3);
    Run(a, 19);
    Run({1, 2, 2, 2, 2, 2, 3}, 2);      // m appears several times: returns one of its occurrences
    return 0;
}
```

### Comparison with linear search

For $a = [1, 2, 2, 2, 2, 2, 3]$ and $m = 2$ linear search returns $1$ (the first occurrence), binary search returns $3$: this is the pseudo-code index, $4$, minus 1. On a sorted array of $n$ elements linear search performs up to $n$ comparisons, binary search at most $\lfloor \log_2 n \rfloor + 1$ iterations.
