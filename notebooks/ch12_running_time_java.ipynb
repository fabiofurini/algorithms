## 1. Find maximum: FindMaximum

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** the maximum value of $a$. In the worst case (increasing array) $S(n) = 4n = \Theta(n)$.

Pseudo-code: section 12.2 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/12-running-time/).

```python
%%writefile FindMaximum.java
public class FindMaximum
{
static long[] cnt = new long[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Line | Executions");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Total steps S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    // FindMaximum(a, n)
    static int findMaximum(int[] a)
    {
        int n = a.length;
        int max = a[0]; cnt[2]++;          // max := a[1]
        int i = 1; cnt[3]++;               // i := 2 in the pseudo-code
        cnt[4]++;                          // the while condition is counted at every evaluation
        while (i < n)                      // i <= n in the pseudo-code
        {
            cnt[5]++;
            if (a[i] > max)
            {
                max = a[i]; cnt[6]++;
            }
            i = i + 1; cnt[8]++;
            cnt[4]++;
        }
        cnt[10]++;
        return max;
    }

    static long stepsOf(int[] a)
    {
        reset();
        findMaximum(a);
        return total();
    }

    public static void main(String[] args)
    {
        // Example of the lecture notes: a[1..4] = (3, 1, 4, 5)
        int[] a = {3, 1, 4, 5};
        int n = a.length;
        reset();
        int m = findMaximum(a);
        System.out.printf("a = %s -> max = %d\n", fmt(a, 0, n), m);
        printSteps(2, 3, 4, 5, 6, 8, 10);
        long t = cnt[6];
        System.out.printf("sum t(i) = %d, S(n) = 3n + 1 + sum t(i) = %d\n", t, 3 * n + 1 + t);

        // Worst case (increasing array) and best case (decreasing array)
        System.out.printf("\n");
        int[] inc = {1, 2, 3, 4};
        System.out.printf("a = %s (increasing): S = %d = 4n = %d\n", fmt(inc, 0, n), stepsOf(inc), 4 * n);
        int[] dec = {4, 3, 2, 1};
        System.out.printf("a = %s (decreasing): S = %d = 3n + 1 = %d\n", fmt(dec, 0, n), stepsOf(dec), 3 * n + 1);

        System.out.printf("\n");
        System.out.printf("Increasing array (1, 2, ..., n):\n");
        System.out.printf("  n | S(n) | 4n\n");
        for (int k = 1; k <= 8; k++)
        {
            int[] b = new int[k];
            for (int i = 0; i < k; i++) b[i] = i + 1;
            System.out.printf("%3d | %4d | %d\n", k, stepsOf(b), 4 * k);
        }
    }
}
```

## 2. Insertion sort: InsertionSort

**Input:** an array $a[1..n]$ of $n$ natural numbers. **Output:** the array $a$ sorted in non-decreasing order. If $t(i)$ is the number of shifts at iteration $i$, then $S(n) = 6n - 4 + 3\sum_{i=2}^{n} t(i)$; in the worst case (decreasing array) $S(n) = \frac{3n^2+9n-8}{2} = \Theta(n^2)$.

Pseudo-code: section 12.3 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/12-running-time/).

```python
%%writefile InsertionSort.java
public class InsertionSort
{
static long[] cnt = new long[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Line | Executions");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Total steps S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    // InsertionSort(a, n)
    static void insertionSort(int[] a, boolean trace)
    {
        int n = a.length;
        long prev = 0;                     // only for the trace
        int i = 1; cnt[2]++;               // i := 2 in the pseudo-code
        cnt[3]++;
        while (i < n)                      // i <= n in the pseudo-code
        {
            int key = a[i]; cnt[4]++;
            int j = i - 1; cnt[5]++;
            cnt[6]++;                      // the while condition is counted at every evaluation
            while (j >= 0 && a[j] > key)   // j >= 1 and a[j] > key in the pseudo-code
            {
                a[j + 1] = a[j]; cnt[7]++;
                j = j - 1; cnt[8]++;
                cnt[6]++;
            }
            a[j + 1] = key; cnt[10]++;
            i = i + 1; cnt[11]++;
            cnt[3]++;
            if (trace)                     // trace: after the increment, i is the pseudo-code index of the iteration just finished
            {
                System.out.println(String.format("%2d | %3d | %4d | %s", i, key, cnt[7] - prev, fmt(a, 0, n)));
                prev = cnt[7];
            }
        }
    }

    static long stepsOf(int[] a)           // works on a copy of a
    {
        reset();
        insertionSort(a.clone(), false);
        return total();
    }

    public static void main(String[] args)
    {
        // Example of the lecture notes: a[1..6] = (5, 2, 4, 6, 1, 3)
        int[] a = {5, 2, 4, 6, 1, 3};
        int n = a.length;
        System.out.printf("a = %s\n", fmt(a, 0, n));
        reset();
        System.out.printf(" i | key | t(i) | array after the iteration\n");
        insertionSort(a, true);
        System.out.printf("sorted: %s\n", fmt(a, 0, n));
        printSteps(2, 3, 4, 5, 6, 7, 8, 10, 11);
        long t = cnt[7];
        System.out.printf("sum t(i) = %d, S(n) = 6n - 4 + 3 sum t(i) = %d\n", t, 6 * n - 4 + 3 * t);

        // Worst case: decreasing array, t(i) = i - 1
        System.out.printf("\n");
        int[] d = {6, 5, 4, 3, 2, 1};
        System.out.printf("a = %s (decreasing): S = %d = (3n^2 + 9n - 8)/2 = %d\n", fmt(d, 0, n), stepsOf(d), (3 * n * n + 9 * n - 8) / 2);
        System.out.printf("sum t(i) = %d = n(n-1)/2 = %d\n", cnt[7], n * (n - 1) / 2);

        System.out.printf("\n");
        System.out.printf("Decreasing array (n, ..., 2, 1):\n");
        System.out.printf("  n | S(n) | (3n^2 + 9n - 8)/2\n");
        for (int k = 1; k <= 8; k++)
        {
            int[] b = new int[k];
            for (int i = 0; i < k; i++) b[i] = k - i;
            System.out.printf("%3d | %4d | %d\n", k, stepsOf(b), (3 * k * k + 9 * k - 8) / 2);
        }
    }
}
```

## 3. Change of base: ChangeToBase2

**Input:** a natural number $n \ge 1$ in base 10. **Output:** the array $d[0..nd-1]$ of binary digits and the number of digits $nd$. For every $n$: $S(n) = 6\lfloor\log_2 n\rfloor + 11 = \Theta(\log_2 n)$.

Pseudo-code: section 12.4 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/12-running-time/).

```python
%%writefile ChangeToBase2.java
public class ChangeToBase2
{
static long[] cnt = new long[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Line | Executions");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Total steps S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    static int floorLog2(int n)            // floor(log2 n) computed with a loop (it does not count as a step)
    {
        int k = 0;
        while (n > 1)
        {
            n = n / 2;
            k = k + 1;
        }
        return k;
    }

    // ChangeToBase2(n): returns the array d[0..nd-1] (nd = d.length)
    static int[] changeToBase2(int n, boolean trace)
    {
        int nd = floorLog2(n) + 1; cnt[2]++;
        int[] d = new int[nd]; cnt[3]++;   // array d[0..nd-1], indexed from 0 as in the pseudo-code
        int k = 0; cnt[4]++;
        cnt[5]++;                          // the while condition is counted at every evaluation
        while (n > 0)
        {
            int q = n / 2; cnt[6]++;
            int r = n % 2; cnt[7]++;
            d[k] = r; cnt[8]++;
            if (trace)                     // trace
            {
                System.out.println(String.format("%10d | %3d | %3d | %3d | d[%d] = %d", k + 1, n, q, r, k, r));
            }
            n = q; cnt[9]++;
            k = k + 1; cnt[10]++;
            cnt[5]++;
        }
        cnt[12]++;
        return d;
    }

    static String digits(int[] d)          // d[nd-1..0] (from the most significant digit)
    {
        StringBuilder s = new StringBuilder();
        for (int k = d.length - 1; k >= 0; k--) s.append(d[k]);
        return s.toString();
    }

    public static void main(String[] args)
    {
        // Example of the lecture notes: n = 13
        int n = 13;
        reset();
        System.out.printf(" iteration |   n |   q |   r | assignment\n");
        int[] d = changeToBase2(n, true);
        System.out.printf("n = %d = (%s)_2: nd = %d, digits d[nd-1..0] = %s\n", n, digits(d), d.length, digits(d));
        printSteps(2, 3, 4, 5, 6, 7, 8, 9, 10, 12);
        System.out.printf("S(n) = 6 floor(log2 n) + 11 = %d\n", 6 * floorLog2(n) + 11);

        // The number of steps grows as log2 n
        System.out.printf("\n");
        System.out.printf("       n | nd | S(n) | 6 floor(log2 n) + 11\n");
        int[] values = {1, 2, 13, 100, 1000, 1000000};
        for (int v : values)
        {
            reset();
            d = changeToBase2(v, false);
            System.out.printf("%8d | %2d | %4d | %d\n", v, d.length, total(), 6 * floorLog2(v) + 11);
        }
    }
}
```

## 4. Primality test: PrimalityTest

**Input:** a natural number $n$. **Output:** `true` if $n$ is prime, `false` otherwise. In the worst case ($n$ prime) $S(n) = 3\lfloor\sqrt{n}\rfloor + 1 = \Theta(\sqrt{n})$, i.e. $2^{\Theta(|I|)}$ as a function of the input size.

Pseudo-code: section 12.5 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/12-running-time/).

```python
%%writefile PrimalityTest.java
public class PrimalityTest
{
static long[] cnt = new long[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Line | Executions");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Total steps S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    static long floorSqrt(long n)          // floor(sqrt(n)) in integer arithmetic
    {
        long r = (long) Math.sqrt((double) n);
        while (r * r > n) r--;
        while ((r + 1) * (r + 1) <= n) r++;
        return r;
    }

    // PrimalityTest(n)
    static boolean primalityTest(long n)
    {
        cnt[2]++;
        if (n < 2)
        {
            cnt[3]++;
            return false;
        }
        long d = 2; cnt[5]++;
        cnt[6]++;                          // the while condition is counted at every evaluation
        while (d <= floorSqrt(n))
        {
            cnt[7]++;
            if (n % d == 0)
            {
                cnt[8]++;
                return false;
            }
            d = d + 1; cnt[10]++;
            cnt[6]++;
        }
        cnt[12]++;
        return true;
    }

    static int bits(long n)                // number of bits of n >= 1
    {
        int k = 0;
        while (n > 0)
        {
            n = n / 2;
            k = k + 1;
        }
        return k;
    }

    public static void main(String[] args)
    {
        // Example of the lecture notes: a prime and a composite number
        long n = 97;
        reset();
        boolean r = primalityTest(n);
        System.out.printf("n = %d: %s\n", n, r ? "true" : "false");
        printSteps(2, 3, 5, 6, 7, 8, 10, 12);
        System.out.printf("S(n) = 3 floor(sqrt(n)) + 1 = %d\n", 3 * floorSqrt(n) + 1);

        System.out.printf("\n");
        n = 91;
        reset();
        r = primalityTest(n);
        long p = 2;
        while (n % p != 0) p = p + 1;
        System.out.printf("n = %d: %s, p = %d\n", n, r ? "true" : "false", p);
        printSteps(2, 3, 5, 6, 7, 8, 10, 12);
        System.out.printf("S(n) = 3p - 1 = %d, while floor(sqrt(n)) = %d\n", 3 * p - 1, floorSqrt(n));

        System.out.printf("\n");
        n = 1;
        reset();
        r = primalityTest(n);
        System.out.printf("n = %d: %s, S = %d\n", n, r ? "true" : "false", total());

        // The worst case are prime numbers: S grows as sqrt(n), i.e. it doubles for every 2 more bits
        System.out.printf("\n");
        System.out.printf("|I| |        n |  S(n) | 3 floor(sqrt(n)) + 1\n");
        long[] primes = {1021, 4093, 16381, 65521, 262139, 1048573, 4194301, 16777213};
        for (long q : primes)
        {
            reset();
            primalityTest(q);
            System.out.printf("%3d | %8d | %5d | %d\n", bits(q), q, total(), 3 * floorSqrt(q) + 1);
        }
    }
}
```
