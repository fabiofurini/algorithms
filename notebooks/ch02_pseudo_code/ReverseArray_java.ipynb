## ReverseArray

**Input:** an array $a[1..n]$ of $n$ numbers. **Output:** the same array with the elements in reversed order.

The element $a[i]$ is exchanged with its symmetric element $a[n-i+1]$; only $\lfloor n/2 \rfloor$ exchanges are needed (with $n$ odd, the middle element stays in place).

```python
%%writefile ReverseArray.java
import java.util.Arrays;

public class ReverseArray {

    // Reverses the elements of a[0..n-1] (in place)
    static void reverseArray(int[] a, int n, boolean trace) {
        if (trace)
            System.out.printf("%3s  %-14s  %s\n", "--", "--", Arrays.toString(a));
        for (int i = 0; i < n / 2; i++) {   // i = 1..⌊n/2⌋ in the pseudo-code
            int tmp = a[i];
            a[i] = a[n - i - 1];            // a[n-i+1] in the pseudo-code
            a[n - i - 1] = tmp;
            if (trace)
                System.out.printf("%3d  %-14s  %s\n", i, "a[" + i + "] <-> a[" + (n - i - 1) + "]", Arrays.toString(a));
        }
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d, %d iterations\n", title, Arrays.toString(a), n, n / 2);
        System.out.printf("%3s  %-14s  %s\n", "i", "exchange", "array a (after the exchange)");
        reverseArray(a, n, true);
        System.out.printf("Return: a = %s\n\n", Arrays.toString(a));
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // “Your turn”: the array comes from the command line
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Your turn", v);
            return;
        }
        run("Instance 1", new int[]{1, 2, 3, 4, 5});
        run("Instance 2", new int[]{10, 20, 30, 40});
    }
}
```
