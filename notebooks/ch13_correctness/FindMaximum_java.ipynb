## Find maximum: FindMaximum

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** the maximum of $a$. **Invariant:** every time the condition is evaluated, $max = \max\{a[1],\dots,a[i-1]\}$.

Pseudo-code: section 13.2 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/13-correctness/).

```python
%%writefile FindMaximum.java
public class FindMaximum
{
// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

static boolean equal(int[] x, int xs, int[] y, int ys, int len)     // x[xs..xs+len-1] == y[ys..ys+len-1]
{
    for (int k = 0; k < len; k++)
        if (x[xs + k] != y[ys + k]) return false;
    return true;
}

    static int maxOf(int[] a, int from, int to)     // maximum of a[from..to-1]
    {
        int m = a[from];
        for (int k = from + 1; k < to; k++)
            if (a[k] > m) m = a[k];
        return m;
    }

    // Invariant: max = maximum of a[1..i-1]
    static void printInvariant(int[] a, int i, int max)
    {
        int m = maxOf(a, 0, i);                     // a[1..i-1] of the pseudo-code
        System.out.println(String.format("%2d | %3d | %-26s | %14d | %s", i + 1, max, fmt(a, 0, i), m, max == m ? "ok" : "FALSE"));
    }

    // FindMaximum(a, n)
    static int findMaximum(int[] a)
    {
        int n = a.length;
        int max = a[0];
        int i = 1;                                  // i := 2
        printInvariant(a, i, max);                  // before the first evaluation of the condition
        while (i < n)                               // i <= n
        {
            if (a[i] > max)
            {
                max = a[i];
            }
            i = i + 1;
            printInvariant(a, i, max);              // after every iteration, i.e. before the next evaluation
        }
        return max;
    }

    public static void main(String[] args)
    {
        int[] a = {9, 12, 3, 16, 19, 11, 12};       // example of the lecture notes
        int n = a.length;
        System.out.printf("a = %s, n = %d\n", fmt(a, 0, n), n);
        System.out.printf(" i | max | a[1..i-1]                  | max{a[1..i-1]} | invariant\n");
        int m = findMaximum(a);
        System.out.printf("The loop ends with i = n + 1 = %d: FindMaximum returns %d\n", n + 1, m);
    }
}
```

**Try it yourself.** Change the array: the invariant holds on every row.

```python
%%writefile ProvaFindMaximum.java
public class ProvaFindMaximum
{
    public static void main(String[] args)
    {
        int[] a = {4, 8, 8, 1, 10, 2};      // <-- change the array
        System.out.printf(" i | max | a[1..i-1]                  | max{a[1..i-1]} | invariant\n");
        int m = FindMaximum.findMaximum(a);
        System.out.printf("returns %d\n", m);
    }
}
```
