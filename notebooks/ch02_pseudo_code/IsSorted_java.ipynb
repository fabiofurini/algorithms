## IsSorted

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** `true` if the values in the array are sorted in non-decreasing order, `false` otherwise.

The `return` statement inside the loop terminates the function as soon as two consecutive elements in the wrong order are found.

```python
%%writefile IsSorted.java
import java.util.Arrays;

public class IsSorted {

    // Returns true if a[0..n-1] is sorted in non-decreasing order
    static boolean isSorted(int[] a, int n, boolean trace) {
        for (int i = 0; i < n - 1; i++) {   // i = 1..n-1 in the pseudo-code
            boolean wrong = a[i] > a[i + 1];    // comparison, saved for the trace
            if (trace)
                System.out.printf("%3d  %4d  %6d  %s\n", i, a[i], a[i + 1], wrong ? "Yes" : "No");
            if (wrong) {
                return false;
            }
        }
        return true;
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s  %4s  %6s  %s\n", "i", "a[i]", "a[i+1]", "a[i]>a[i+1]?");
        boolean result = isSorted(a, n, true);
        System.out.printf("Return: %s\n\n", result ? "true" : "false");
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // “Your turn”: the array comes from the command line
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Your turn", v);
            return;
        }
        run("Instance 1", new int[]{2, 5, 7, 9});
        run("Instance 2", new int[]{3, 1, 4, 2});
    }
}
```
