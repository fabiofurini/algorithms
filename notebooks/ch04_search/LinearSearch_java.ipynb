## LinearSearch

**Input:** an array $a[1..n]$ of $n$ natural numbers and a target value $m$. **Output:** the index of $m$ in $a$, or $-1$ if $m$ is not found.

The algorithm compares $m$ with the elements of the array one after the other and stops at the first one equal to $m$: the returned index is the smallest $k$ with $a[k]=m$. In the code the indices start from 0, so the returned index is the one of the pseudo-code minus 1.

```python
%%writefile LinearSearch.java
import java.util.Arrays;

public class LinearSearch {

    // Returns the index of m in a[0..n-1], or -1 if m is not found
    static int linearSearch(int[] a, int n, int m, boolean trace) {
        for (int i = 0; i < n; i++) {   // i = 1..n in the pseudo-code
            if (a[i] == m) {
                if (trace)
                    System.out.printf("%10d %3d  %-16s  %s\n", i + 1, i, "a[" + i + "] = " + a[i] + " = m", "Found! Return " + i);
                return i;               // element found at index i
            }
            if (trace)
                System.out.printf("%10d %3d  %-16s  %s\n", i + 1, i, "a[" + i + "] = " + a[i] + " != m", "continue");
        }
        if (trace)
            System.out.printf("%10s %3d  %-16s  %s\n", "-", n, "-", "i = n: the loop ends, not found! Return -1");
        return -1;                      // element not found
    }

    static void run(int[] a, int m) {
        int n = a.length;
        System.out.printf("Search: m = %d, a = %s, n = %d\n", m, Arrays.toString(a), n);
        System.out.printf("%10s %3s  %-16s  %s\n", "iteration", "i", "comparison", "action");
        int result = linearSearch(a, n, m, true);
        System.out.printf("Return: %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 1) {          // “Your turn”: first number = m, the others = array
            int[] v = new int[args.length - 1];
            for (int k = 1; k < args.length; k++) v[k - 1] = Integer.parseInt(args[k]);
            run(v, Integer.parseInt(args[0]));
            return;
        }
        int[] a = {4, 2, 7, 1, 5, 9, 3};
        run(a, 5);
        run(a, 6);
        run(new int[]{1, 2, 2, 2, 2, 2, 3}, 2);     // m appears several times: returns the first index
    }
}
```
