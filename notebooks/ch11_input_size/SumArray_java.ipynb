## Sum of the entries of an array: SumArray

**Input:** an array $a[1..n]$ of $n \ge 1$ natural numbers. **Output:** $a[1]+a[2]+\cdots+a[n]$. Number of steps: $S(n) = 3n + 4$, whatever the values.

Pseudo-code: section 11.6 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/11-input-size/).

```python
%%writefile SumArray.java
public class SumArray
{
static long[] cnt = new long[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

static void reset()
{
    java.util.Arrays.fill(cnt, 0);
}

static long total()
{
    long s = 0;
    for (long c : cnt) s += c;
    return s;
}

static void printSteps(int... lines)
{
    System.out.println("Line | Executions");
    for (int r : lines) System.out.println(String.format("%4d | %10d", r, cnt[r]));
    System.out.println("Total steps S = " + total());
}

// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

    // SumArray(a, n)
    static int sumArray(int[] a)
    {
        int n = a.length;
        int s = 0; cnt[2]++;
        int i = 0; cnt[3]++;               // i := 1 in the pseudo-code
        cnt[4]++;                          // the while condition is counted at every evaluation
        while (i < n)                      // i <= n in the pseudo-code
        {
            s = s + a[i]; cnt[5]++;
            i = i + 1; cnt[6]++;
            cnt[4]++;
        }
        cnt[8]++;
        return s;
    }

    public static void main(String[] args)
    {
        int[] a = {6, 1, 3, 8};
        int n = a.length;
        reset();
        int s = sumArray(a);
        System.out.printf("a = %s, n = %d, sum = %d\n", fmt(a, 0, n), n, s);
        printSteps(2, 3, 4, 5, 6, 8);
        System.out.printf("Formula: S(n) = 3n + 4 = %d\n", 3 * n + 4);

        // Worst case: the array of |I| ones has input size |I| and takes exactly 3|I| + 4 steps
        System.out.printf("\n");
        System.out.printf("Array of |I| ones:\n");
        int[] sizes = {4, 8, 16, 32};
        for (int size : sizes)
        {
            int[] ones = new int[size];
            java.util.Arrays.fill(ones, 1);
            reset();
            sumArray(ones);
            System.out.printf("|I| = %2d: S = %3d = 3|I| + 4 = %d\n", size, total(), 3 * size + 4);
        }
    }
}
```

**Try it yourself.** Change the array and check that the steps are always $3n + 4$.

```python
%%writefile ProvaSumArray.java
public class ProvaSumArray
{
    public static void main(String[] args)
    {
        int[] a = {5, 1, 9, 2, 7};      // <-- change the array
        SumArray.reset();
        int s = SumArray.sumArray(a);
        System.out.printf("sum = %d, S = %d, 3n + 4 = %d\n", s, SumArray.total(), 3 * a.length + 4);
    }
}
```
