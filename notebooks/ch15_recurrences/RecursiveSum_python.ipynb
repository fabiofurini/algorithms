## RecursiveSum and the count of the steps

**RecursiveSum**$(a,n)$ — *Input*: an array $a[1..n]$ of $n\ge1$ integers. *Output*: the sum $a[1]+\dots+a[n]$.

The code follows the pseudo-code (lines 2, 3, 5, 6); a counter `steps` adds up the steps executed by every call, as in the chapter: 2 steps if $n=1$, 3 steps plus the recursive call if $n>1$. The second function is `sumRecDecrease` of the code of the course, which stops at the empty array ($n=0$) and therefore makes one more call.

```python
steps = 0   # steps executed (lines of pseudo-code)
calls = 0   # number of calls

def recursive_sum(a, n):
    """Sum of a[0..n-1] (pseudo-code: a[1..n]), n >= 1."""
    global steps, calls
    calls += 1
    steps += 1                    # line 2: if n = 1
    if n == 1:
        steps += 1                # line 3: return a[1]
        return a[0]
    steps += 2                    # lines 5 and 6
    s = recursive_sum(a, n - 1)
    return s + a[n - 1]

def sum_rec_decrease(a, n):
    """Variant of the course code: base case n = 0."""
    global calls
    calls += 1
    if n == 0:
        return 0
    return sum_rec_decrease(a, n - 1) + a[n - 1]

a = [3, 1, 4, 1, 5, 9, 2, 6]
print("a =", a)
calls = 0
print("RecursiveSum(a, 8) =", recursive_sum(a, len(a)), "with", calls, "calls")
calls = 0
print("sumRecDecrease(a, 8) =", sum_rec_decrease(a, len(a)), "with", calls, "calls")
print("n  measured S(n)  3n-1")
for n in range(1, 9):
    steps = 0
    recursive_sum(a, n)
    print(f"{n:<2d} {steps:>12d} {3 * n - 1:>5d}")
```

**Try it yourself.** Change the array `a` (also its length) and check that the measured steps are always $3n-1$.

```python
# Change the array and check that the measured steps are always 3n-1.
a = [7, -2, 5, 0, 11]
steps = 0
print(recursive_sum(a, len(a)), steps, 3 * len(a) - 1)
```
