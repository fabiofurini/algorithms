## Euclidean algorithm: EuclideanAlgorithm

**Input:** two natural numbers $n, m \ge 1$. **Output:** $\gcd(n, m)$. **Invariant:** $n \ge 1$, $m \ge 0$ and $\gcd(n,m) = \gcd(n_0,m_0)$, where $n_0, m_0$ are the input values.

Pseudo-code: section 13.5 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/13-correctness/).

```python
%%writefile EuclideanAlgorithm.java
public class EuclideanAlgorithm
{
// "[a[from], ..., a[to-1]]"
static String fmt(int[] a, int from, int to)
{
    StringBuilder s = new StringBuilder("[");
    for (int i = from; i < to; i++)
    {
        if (i > from) s.append(", ");
        s.append(a[i]);
    }
    return s.append("]").toString();
}

static boolean equal(int[] x, int xs, int[] y, int ys, int len)     // x[xs..xs+len-1] == y[ys..ys+len-1]
{
    for (int k = 0; k < len; k++)
        if (x[xs + k] != y[ys + k]) return false;
    return true;
}

    static int gcdCheck(int n, int m)              // gcd by enumeration of the divisors, independent of the algorithm
    {
        if (m == 0) return n;
        int d = Math.min(n, m);
        while (n % d != 0 || m % d != 0) d = d - 1;
        return d;
    }

    // Invariant: n >= 1, m >= 0 and gcd(n, m) = gcd(n0, m0)
    static void printRow(int n, int m, int g0)
    {
        int g = gcdCheck(n, m);
        boolean ok = (n >= 1 && m >= 0 && g == g0);
        System.out.println(String.format("%4d | %4d | %9d | %s", n, m, g, ok ? "ok" : "FALSE"));
    }

    // EuclideanAlgorithm(n, m)
    static int euclideanAlgorithm(int n, int m)
    {
        int g0 = gcdCheck(n, m);                   // gcd(n0, m0), only to check the invariant
        printRow(n, m, g0);
        while (m != 0)
        {
            int r = n % m;
            n = m;
            m = r;
            printRow(n, m, g0);
        }
        return n;
    }

    public static void main(String[] args)
    {
        int n = 252, m = 105;      // example of the lecture notes
        System.out.printf("(n0, m0) = (%d, %d)\n", n, m);
        System.out.printf("   n |    m | gcd(n, m) | invariant\n");
        int g = euclideanAlgorithm(n, m);
        System.out.printf("The loop ends with m = 0: it returns n = %d\n", g);

        System.out.printf("\n");
        n = 48; m = 18;
        System.out.printf("(n0, m0) = (%d, %d)\n", n, m);
        System.out.printf("   n |    m | gcd(n, m) | invariant\n");
        g = euclideanAlgorithm(n, m);
        System.out.printf("The loop ends with m = 0: it returns n = %d\n", g);
    }
}
```

**Try it yourself.** Change the two numbers: the $\gcd$ stays constant on every row.

```python
%%writefile ProvaEuclideanAlgorithm.java
public class ProvaEuclideanAlgorithm
{
    public static void main(String[] args)
    {
        int n = 1071, m = 462;      // <-- change the two numbers
        System.out.printf("   n |    m | gcd(n, m) | invariant\n");
        int g = EuclideanAlgorithm.euclideanAlgorithm(n, m);
        System.out.printf("returns %d\n", g);
    }
}
```
