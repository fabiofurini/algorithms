## Sum of the entries of an array: SumArray

**Input:** an array $a[1..n]$ of $n \ge 1$ natural numbers. **Output:** $a[1]+a[2]+\cdots+a[n]$. Number of steps: $S(n) = 3n + 4$, whatever the values.

Pseudo-code: section 11.6 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/11-input-size/).

```python
%%writefile sum_array.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Line | Executions\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Total steps S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// SumArray(a, n)
int SumArray(int a[], int n)
{
    int s = 0; cnt[2]++;
    int i = 0; cnt[3]++;               // i := 1 in the pseudo-code
    cnt[4]++;                          // the while condition is counted at every evaluation
    while (i < n)                      // i <= n in the pseudo-code
    {
        s = s + a[i]; cnt[5]++;
        i = i + 1; cnt[6]++;
        cnt[4]++;
    }
    cnt[8]++;
    return s;
}

#ifndef SOLO_LIBRERIA
int main()
{
    int a[] = {6, 1, 3, 8};
    int n = 4;
    Reset();
    int s = SumArray(a, n);
    printf("a = %s, n = %d, sum = %d\n", Fmt(a, 0, n).c_str(), n, s);
    PrintSteps({2, 3, 4, 5, 6, 8});
    printf("Formula: S(n) = 3n + 4 = %d\n", 3 * n + 4);

    // Worst case: the array of |I| ones has input size |I| and takes exactly 3|I| + 4 steps
    printf("\n");
    printf("Array of |I| ones:\n");
    int sizes[] = {4, 8, 16, 32};
    for (int k = 0; k < 4; k++)
    {
        int size = sizes[k];
        int ones[32];
        for (int i = 0; i < size; i++) ones[i] = 1;
        Reset();
        SumArray(ones, size);
        printf("|I| = %2d: S = %3lld = 3|I| + 4 = %d\n", size, Total(), 3 * size + 4);
    }
    return 0;
}
#endif
```

```python
!g++ -std=c++17 -O2 -o sum_array sum_array.cpp && ./sum_array
```

**Try it yourself.** Change the array and check that the steps are always $3n + 4$.

```python
%%writefile prova_sum_array.cpp
#define SOLO_LIBRERIA   // do not run the main of sum_array.cpp
#include "sum_array.cpp"

int main()
{
    int a[] = {5, 1, 9, 2, 7};      // <-- change the array
    int n = 5;
    Reset();
    int s = SumArray(a, n);
    printf("sum = %d, S = %lld, 3n + 4 = %d\n", s, Total(), 3 * n + 4);
    return 0;
}
```
