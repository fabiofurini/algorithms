## FindMaximum

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** the maximum value in the array.

The variable `max` starts from the first element and is updated whenever a larger element is found.

```python
%%writefile FindMaximum.java
import java.util.Arrays;

public class FindMaximum {

    // Returns the maximum of a[0..n-1], with n >= 1
    static int findMaximum(int[] a, int n, boolean trace) {
        int max = a[0];                 // start from the first element
        if (trace)
            System.out.printf("%3s  %4s  %4d  %s\n", "--", "--", max, "--");
        for (int i = 1; i < n; i++) {   // i = 2..n in the pseudo-code
            boolean bigger = a[i] > max;    // comparison, saved for the trace
            if (bigger) {
                max = a[i];             // update the maximum
            }
            if (trace)
                System.out.printf("%3d  %4d  %4d  %s\n", i, a[i], max, bigger ? "Yes" : "No");
        }
        return max;
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s  %4s  %4s  %s\n", "i", "a[i]", "max", "a[i]>max?");
        int result = findMaximum(a, n, true);
        System.out.printf("Return: max = %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // “Your turn”: the array comes from the command line
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Your turn", v);
            return;
        }
        run("Instance 1", new int[]{9, 12, 3, 16, 19, 11, 12});
        run("Instance 2", new int[]{3, 7, 2, 9, 1});
    }
}
```

## CountPositive

**Input:** an array $a[1..n]$ of $n$ integer numbers. **Output:** the number of positive elements in the array.

```python
%%writefile CountPositive.java
import java.util.Arrays;

public class CountPositive {

    // Returns the number of positive elements of a[0..n-1]
    static int countPositive(int[] a, int n, boolean trace) {
        int count = 0;
        if (trace)
            System.out.printf("%3s  %4s  %5d  %s\n", "--", "--", count, "--");
        for (int i = 0; i < n; i++) {   // i = 1..n in the pseudo-code
            boolean positive = a[i] > 0;    // comparison, saved for the trace
            if (positive) {
                count = count + 1;
            }
            if (trace)
                System.out.printf("%3d  %4d  %5d  %s\n", i, a[i], count, positive ? "Yes" : "No");
        }
        return count;
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s  %4s  %5s  %s\n", "i", "a[i]", "count", "a[i]>0?");
        int result = countPositive(a, n, true);
        System.out.printf("Return: count = %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // “Your turn”: the array comes from the command line
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Your turn", v);
            return;
        }
        run("Instance 1", new int[]{-2, 5, -1, 3, 0});
        run("Instance 2", new int[]{4, -3, 7, 2});
    }
}
```

## IsSorted

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** `true` if the values in the array are sorted in non-decreasing order, `false` otherwise.

The `return` statement inside the loop terminates the function as soon as two consecutive elements in the wrong order are found.

```python
%%writefile IsSorted.java
import java.util.Arrays;

public class IsSorted {

    // Returns true if a[0..n-1] is sorted in non-decreasing order
    static boolean isSorted(int[] a, int n, boolean trace) {
        for (int i = 0; i < n - 1; i++) {   // i = 1..n-1 in the pseudo-code
            boolean wrong = a[i] > a[i + 1];    // comparison, saved for the trace
            if (trace)
                System.out.printf("%3d  %4d  %6d  %s\n", i, a[i], a[i + 1], wrong ? "Yes" : "No");
            if (wrong) {
                return false;
            }
        }
        return true;
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d\n", title, Arrays.toString(a), n);
        System.out.printf("%3s  %4s  %6s  %s\n", "i", "a[i]", "a[i+1]", "a[i]>a[i+1]?");
        boolean result = isSorted(a, n, true);
        System.out.printf("Return: %s\n\n", result ? "true" : "false");
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // “Your turn”: the array comes from the command line
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Your turn", v);
            return;
        }
        run("Instance 1", new int[]{2, 5, 7, 9});
        run("Instance 2", new int[]{3, 1, 4, 2});
    }
}
```

## ScalarProduct

**Input:** two arrays $a[1..n]$ and $b[1..n]$ of size $n \ge 1$. **Output:** the scalar product $\sum_{i=1}^{n} a[i] \cdot b[i]$ of the two arrays.

```python
%%writefile ScalarProduct.java
import java.util.Arrays;

public class ScalarProduct {

    // Returns the scalar product of a[0..n-1] and b[0..n-1], with n >= 1
    static int scalarProduct(int[] a, int[] b, int n, boolean trace) {
        int prod = a[0] * b[0];
        if (trace)
            System.out.printf("%3s  %4s  %4s  %9s  %5d\n", "--", "--", "--", "--", prod);
        for (int i = 1; i < n; i++) {   // i = 2..n in the pseudo-code
            prod = prod + a[i] * b[i];
            if (trace)
                System.out.printf("%3d  %4d  %4d  %9d  %5d\n", i, a[i], b[i], a[i] * b[i], prod);
        }
        return prod;
    }

    static void run(String title, int[] a, int[] b) {
        int n = a.length;
        System.out.printf("%s: a = %s, b = %s, n = %d\n", title, Arrays.toString(a), Arrays.toString(b), n);
        System.out.printf("%3s  %4s  %4s  %9s  %5s\n", "i", "a[i]", "b[i]", "a[i]*b[i]", "prod");
        int result = scalarProduct(a, b, n, true);
        System.out.printf("Return: prod = %d\n\n", result);
    }

    public static void main(String[] args) {
        if (args.length > 1) {          // “Your turn”: the numbers come from the command line (half for a, half for b)
            int n = args.length / 2;
            int[] a = new int[n], b = new int[n];
            for (int k = 0; k < n; k++) {
                a[k] = Integer.parseInt(args[k]);
                b[k] = Integer.parseInt(args[n + k]);
            }
            run("Your turn", a, b);
            return;
        }
        run("Instance 1", new int[]{1, 2, 3}, new int[]{4, 5, 6});
        run("Instance 2", new int[]{2, 0, -1}, new int[]{3, 4, 5});
    }
}
```

## ReverseArray

**Input:** an array $a[1..n]$ of $n$ numbers. **Output:** the same array with the elements in reversed order.

The element $a[i]$ is exchanged with its symmetric element $a[n-i+1]$; only $\lfloor n/2 \rfloor$ exchanges are needed (with $n$ odd, the middle element stays in place).

```python
%%writefile ReverseArray.java
import java.util.Arrays;

public class ReverseArray {

    // Reverses the elements of a[0..n-1] (in place)
    static void reverseArray(int[] a, int n, boolean trace) {
        if (trace)
            System.out.printf("%3s  %-14s  %s\n", "--", "--", Arrays.toString(a));
        for (int i = 0; i < n / 2; i++) {   // i = 1..⌊n/2⌋ in the pseudo-code
            int tmp = a[i];
            a[i] = a[n - i - 1];            // a[n-i+1] in the pseudo-code
            a[n - i - 1] = tmp;
            if (trace)
                System.out.printf("%3d  %-14s  %s\n", i, "a[" + i + "] <-> a[" + (n - i - 1) + "]", Arrays.toString(a));
        }
    }

    static void run(String title, int[] a) {
        int n = a.length;
        System.out.printf("%s: a = %s, n = %d, %d iterations\n", title, Arrays.toString(a), n, n / 2);
        System.out.printf("%3s  %-14s  %s\n", "i", "exchange", "array a (after the exchange)");
        reverseArray(a, n, true);
        System.out.printf("Return: a = %s\n\n", Arrays.toString(a));
    }

    public static void main(String[] args) {
        if (args.length > 0) {          // “Your turn”: the array comes from the command line
            int[] v = new int[args.length];
            for (int k = 0; k < args.length; k++) v[k] = Integer.parseInt(args[k]);
            run("Your turn", v);
            return;
        }
        run("Instance 1", new int[]{1, 2, 3, 4, 5});
        run("Instance 2", new int[]{10, 20, 30, 40});
    }
}
```
