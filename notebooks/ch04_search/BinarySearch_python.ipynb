## BinarySearch

**Input:** an array $a[1..n]$ of $n$ natural numbers **sorted in non-decreasing order** and a target value $m$. **Output:** the index of $m$ in $a$, or $-1$ if $m$ is not found.

The algorithm keeps the search interval $a[left..right]$ and at each iteration compares $m$ with the middle element $a[mid]$, halving the interval: it executes at most $\lfloor \log_2 n \rfloor + 1$ iterations. If $m$ appears more than once it returns the index of one of its occurrences, not necessarily the first one.

```python
def binary_search(a, n, m, trace=False):
    left = 0                            # left boundary (1 in the pseudo-code)
    right = n - 1                       # right boundary (n in the pseudo-code)
    iteration = 0                       # only for the trace
    while left <= right:
        iteration += 1
        mid = left + (right - left) // 2    # middle index
        if a[mid] == m:
            if trace:
                print("%10d %5d %5d %4d  %s" % (iteration, left, right, mid, "a[%d] = %d = m => Found! Return %d" % (mid, a[mid], mid)))
            return mid                  # element found
        if a[mid] < m:
            if trace:
                print("%10d %5d %5d %4d  %s" % (iteration, left, right, mid, "a[%d] = %d < %d => left = %d + 1 = %d" % (mid, a[mid], m, mid, mid + 1)))
            left = mid + 1              # search in the right half
        else:
            if trace:
                print("%10d %5d %5d %4d  %s" % (iteration, left, right, mid, "a[%d] = %d > %d => right = %d - 1 = %d" % (mid, a[mid], m, mid, mid - 1)))
            right = mid - 1             # search in the left half
    if trace:
        print("%10s %5d %5d %4s  %s" % ("-", left, right, "-", "left > right => Not found! Return -1"))
    return -1                           # element not found
```

### Execution on the examples of the chapter

```python
def run(a, m):
    n = len(a)
    print("Search: m = %d, a = %s, n = %d" % (m, a, n))
    print("%10s %5s %5s %4s  %s" % ("iteration", "left", "right", "mid", "action"))
    result = binary_search(a, n, m, trace=True)
    print("Return: %d" % result)
    print()


a = [3, 9, 11, 12, 12, 16, 19]
run(a, 12)
run(a, 10)
run(a, 3)
run(a, 19)
run([1, 2, 2, 2, 2, 2, 3], 2)    # m appears several times: returns one of its occurrences
```
