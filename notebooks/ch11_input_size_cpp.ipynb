## 1. The input size of the examples

**Input:** a natural number, two natural numbers or an array of natural numbers. **Output:** the input size $|I|$, i.e. the number of bits needed to write it in binary ($\lfloor \log_2 n \rfloor + 1$ bits for a number $n \ge 1$).

See: sections 11.1, 11.3, 11.4, 11.5 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/11-input-size/).

```python
%%writefile dimensione_input_esempi.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Line | Executions\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Total steps S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// Number of bits of a natural n >= 1: floor(log2 n) + 1
int Bits(int n)
{
    int k = 0;
    while (n > 0)
    {
        n = n / 2;        // halve until reaching 0
        k = k + 1;
    }
    return k;
}

// Digits of n in base 2 (as a string)
string Binary(int n)
{
    string s = "";
    while (n > 0)
    {
        s = to_string(n % 2) + s;
        n = n / 2;
    }
    return s;
}

#ifndef SOLO_LIBRERIA
int main()
{
    // Example of the lecture notes: the number of bits of some instances
    int n = 37, m = 200;
    printf("One number: n = %d = (%s)_2, |I| = %d\n", n, Binary(n).c_str(), Bits(n));
    printf("Two numbers: n = %d, m = %d = (%s)_2, |I| = %d + %d = %d\n", n, m, Binary(m).c_str(), Bits(n), Bits(m), Bits(n) + Bits(m));
    int a[] = {6, 1, 3, 8};
    int b[4], sum = 0, M = 0;
    for (int i = 0; i < 4; i++)
    {
        b[i] = Bits(a[i]);
        sum += b[i];
        if (a[i] > M) M = a[i];
    }
    printf("One array: a = %s, bits per entry = %s, |I| = %d\n", Fmt(a, 0, 4).c_str(), Fmt(b, 0, 4).c_str(), sum);

    // The bounds of the lecture notes (values rounded to two decimals)
    printf("\n");
    printf("One number: log2(n) = %.2f <= |I| = %d <= 2 log2(n) = %.2f\n", log2(n), Bits(n), 2 * log2(n));
    double s = log2(n) + log2(m);
    printf("Two numbers: log2(n) + log2(m) = %.2f <= |I| = %d <= 2 (log2(n) + log2(m)) = %.2f\n", s, Bits(n) + Bits(m), 2 * s);
    int mx = n > m ? n : m;
    printf("Two numbers and maximum: log2(max) = %.2f <= |I| = %d <= 4 log2(max) = %.2f\n", log2(mx), Bits(n) + Bits(m), 4 * log2(mx));
    printf("One array: n = %d <= |I| = %d <= n (1 + log2 M) = %.2f\n", 4, sum, 4 * (1 + log2(M)));

    // The upper bound is not tight: a[1] = M = 2^n, a[2..n] = 1 has |I| = 2n, but n(1 + log2 M) = n(n + 1)
    printf("\n");
    printf("   n |      M |  |I| = 2n | n (1 + log2 M)\n");
    for (int k = 1; k <= 10; k++)
    {
        int big = 1 << k;
        int size = Bits(big) + (k - 1);     // a[1] = M, a[2..n] = 1 (1 bit each)
        printf("%4d | %6d | %9d | %13d\n", k, big, size, k * (k + 1));
    }
    return 0;
}
#endif
```

## 2. Sum of the entries of an array: SumArray

**Input:** an array $a[1..n]$ of $n \ge 1$ natural numbers. **Output:** $a[1]+a[2]+\cdots+a[n]$. Number of steps: $S(n) = 3n + 4$, whatever the values.

Pseudo-code: section 11.6 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis/11-input-size/).

```python
%%writefile sum_array.cpp
#include <cstdio>
#include <cmath>
#include <initializer_list>
#include <string>
using namespace std;

long long cnt[20];   // cnt[r] = number of executions of line r (numbering of the lecture notes)

void Reset()
{
    for (int r = 0; r < 20; r++) cnt[r] = 0;
}

long long Total()
{
    long long s = 0;
    for (int r = 0; r < 20; r++) s += cnt[r];
    return s;
}

void PrintSteps(initializer_list<int> lines)
{
    printf("Line | Executions\n");
    for (int r : lines) printf("%4d | %10lld\n", r, cnt[r]);
    printf("Total steps S = %lld\n", Total());
}

// "[a[from], ..., a[to-1]]"
string Fmt(const int* a, int from, int to)
{
    string s = "[";
    for (int i = from; i < to; i++)
    {
        if (i > from) s += ", ";
        s += to_string(a[i]);
    }
    return s + "]";
}

// SumArray(a, n)
int SumArray(int a[], int n)
{
    int s = 0; cnt[2]++;
    int i = 0; cnt[3]++;               // i := 1 in the pseudo-code
    cnt[4]++;                          // the while condition is counted at every evaluation
    while (i < n)                      // i <= n in the pseudo-code
    {
        s = s + a[i]; cnt[5]++;
        i = i + 1; cnt[6]++;
        cnt[4]++;
    }
    cnt[8]++;
    return s;
}

#ifndef SOLO_LIBRERIA
int main()
{
    int a[] = {6, 1, 3, 8};
    int n = 4;
    Reset();
    int s = SumArray(a, n);
    printf("a = %s, n = %d, sum = %d\n", Fmt(a, 0, n).c_str(), n, s);
    PrintSteps({2, 3, 4, 5, 6, 8});
    printf("Formula: S(n) = 3n + 4 = %d\n", 3 * n + 4);

    // Worst case: the array of |I| ones has input size |I| and takes exactly 3|I| + 4 steps
    printf("\n");
    printf("Array of |I| ones:\n");
    int sizes[] = {4, 8, 16, 32};
    for (int k = 0; k < 4; k++)
    {
        int size = sizes[k];
        int ones[32];
        for (int i = 0; i < size; i++) ones[i] = 1;
        Reset();
        SumArray(ones, size);
        printf("|I| = %2d: S = %3lld = 3|I| + 4 = %d\n", size, Total(), 3 * size + 4);
    }
    return 0;
}
#endif
```

```python
%%writefile prova_sum_array.cpp
#define SOLO_LIBRERIA   // do not run the main of sum_array.cpp
#include "sum_array.cpp"

int main()
{
    int a[] = {5, 1, 9, 2, 7};      // <-- change the array
    int n = 5;
    Reset();
    int s = SumArray(a, n);
    printf("sum = %d, S = %lld, 3n + 4 = %d\n", s, Total(), 3 * n + 4);
    return 0;
}
```
