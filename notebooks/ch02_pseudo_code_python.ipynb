## FindMaximum

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** the maximum value in the array.

The variable `max` starts from the first element and is updated whenever a larger element is found.

```python
def find_maximum(a, n, trace=False):
    max_val = a[0]                  # start from the first element
    if trace:
        print("%3s  %4s  %4d  %s" % ("--", "--", max_val, "--"))
    for i in range(1, n):           # i = 2..n in the pseudo-code
        bigger = a[i] > max_val     # comparison, saved for the trace
        if bigger:
            max_val = a[i]          # update the maximum
        if trace:
            print("%3d  %4d  %4d  %s" % (i, a[i], max_val, "Yes" if bigger else "No"))
    return max_val
```

### Execution on the examples of the chapter

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s  %4s  %4s  %s" % ("i", "a[i]", "max", "a[i]>max?"))
    result = find_maximum(a, n, trace=True)
    print("Return: max = %d" % result)
    print()


run("Instance 1", [9, 12, 3, 16, 19, 11, 12])
run("Instance 2", [3, 7, 2, 9, 1])
```

## CountPositive

**Input:** an array $a[1..n]$ of $n$ integer numbers. **Output:** the number of positive elements in the array.

```python
def count_positive(a, n, trace=False):
    count = 0
    if trace:
        print("%3s  %4s  %5s  %s" % ("--", "--", count, "--"))
    for i in range(n):              # i = 1..n in the pseudo-code
        positive = a[i] > 0         # comparison, saved for the trace
        if positive:
            count = count + 1
        if trace:
            print("%3d  %4d  %5d  %s" % (i, a[i], count, "Yes" if positive else "No"))
    return count
```

### Execution on the examples of the chapter

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s  %4s  %5s  %s" % ("i", "a[i]", "count", "a[i]>0?"))
    result = count_positive(a, n, trace=True)
    print("Return: count = %d" % result)
    print()


run("Instance 1", [-2, 5, -1, 3, 0])
run("Instance 2", [4, -3, 7, 2])
```

## IsSorted

**Input:** an array $a[1..n]$ of $n \ge 1$ integer numbers. **Output:** `true` if the values in the array are sorted in non-decreasing order, `false` otherwise.

The `return` statement inside the loop terminates the function as soon as two consecutive elements in the wrong order are found.

```python
def is_sorted(a, n, trace=False):
    for i in range(n - 1):          # i = 1..n-1 in the pseudo-code
        wrong = a[i] > a[i + 1]     # comparison, saved for the trace
        if trace:
            print("%3d  %4d  %6d  %s" % (i, a[i], a[i + 1], "Yes" if wrong else "No"))
        if wrong:
            return False
    return True
```

### Execution on the examples of the chapter

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d" % (title, a, n))
    print("%3s  %4s  %6s  %s" % ("i", "a[i]", "a[i+1]", "a[i]>a[i+1]?"))
    result = is_sorted(a, n, trace=True)
    print("Return: %s" % str(result).lower())
    print()


run("Instance 1", [2, 5, 7, 9])
run("Instance 2", [3, 1, 4, 2])
```

## ScalarProduct

**Input:** two arrays $a[1..n]$ and $b[1..n]$ of size $n \ge 1$. **Output:** the scalar product $\sum_{i=1}^{n} a[i] \cdot b[i]$ of the two arrays.

```python
def scalar_product(a, b, n, trace=False):
    prod = a[0] * b[0]
    if trace:
        print("%3s  %4s  %4s  %9s  %5d" % ("--", "--", "--", "--", prod))
    for i in range(1, n):           # i = 2..n in the pseudo-code
        prod = prod + a[i] * b[i]
        if trace:
            print("%3d  %4d  %4d  %9d  %5d" % (i, a[i], b[i], a[i] * b[i], prod))
    return prod
```

### Execution on the examples of the chapter

```python
def run(title, a, b):
    n = len(a)
    print("%s: a = %s, b = %s, n = %d" % (title, a, b, n))
    print("%3s  %4s  %4s  %9s  %5s" % ("i", "a[i]", "b[i]", "a[i]*b[i]", "prod"))
    result = scalar_product(a, b, n, trace=True)
    print("Return: prod = %d" % result)
    print()


run("Instance 1", [1, 2, 3], [4, 5, 6])
run("Instance 2", [2, 0, -1], [3, 4, 5])
```

## ReverseArray

**Input:** an array $a[1..n]$ of $n$ numbers. **Output:** the same array with the elements in reversed order.

The element $a[i]$ is exchanged with its symmetric element $a[n-i+1]$; only $\lfloor n/2 \rfloor$ exchanges are needed (with $n$ odd, the middle element stays in place).

```python
def reverse_array(a, n, trace=False):
    if trace:
        print("%3s  %-14s  %s" % ("--", "--", a))
    for i in range(n // 2):         # i = 1..⌊n/2⌋ in the pseudo-code
        tmp = a[i]
        a[i] = a[n - i - 1]         # a[n-i+1] in the pseudo-code
        a[n - i - 1] = tmp
        if trace:
            print("%3d  %-14s  %s" % (i, "a[%d] <-> a[%d]" % (i, n - i - 1), a))
    return a
```

### Execution on the examples of the chapter

```python
def run(title, a):
    n = len(a)
    print("%s: a = %s, n = %d, %d iterations" % (title, a, n, n // 2))
    print("%3s  %-14s  %s" % ("i", "exchange", "array a (after the exchange)"))
    result = reverse_array(a, n, trace=True)
    print("Return: a = %s" % result)
    print()


run("Instance 1", [1, 2, 3, 4, 5])
run("Instance 2", [10, 20, 30, 40])
```
