## Tools: the step counter

The cell defines the array `cnt` (one entry per line of the pseudo-code), `reset()` to clear it and `print_steps()` to print the table of executions.

```python
cnt = [0] * 20      # cnt[r] = number of executions of line r (lines numbered as in the lecture notes)


def reset():
    for r in range(len(cnt)):
        cnt[r] = 0


def print_steps(*lines):
    print("Line | Executions")
    for r in lines:
        print("%4d | %10d" % (r, cnt[r]))
    print("Total steps S = %d" % sum(cnt))
```

## 1. The input size of the examples

**Input:** a natural number, two natural numbers or an array of natural numbers. **Output:** the input size $|I|$, i.e. the number of bits needed to write it in binary ($\lfloor \log_2 n \rfloor + 1$ bits for a number $n \ge 1$).

See: sections 11.1, 11.3, 11.4, 11.5 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis-recursion/11-input-size/).

```python
def bits(n):
    """Number of bits of a natural n >= 1: floor(log2 n) + 1."""
    k = 0
    while n > 0:
        n = n // 2        # halve until reaching 0
        k = k + 1
    return k


def binary(n):
    """Digits of n in base 2 (as a string)."""
    s = ""
    while n > 0:
        s = str(n % 2) + s
        n = n // 2
    return s
```

```python
from math import log2

# Example of the lecture notes: the number of bits of some instances
n = 37
print("One number: n = %d = (%s)_2, |I| = %d" % (n, binary(n), bits(n)))
m = 200
print("Two numbers: n = %d, m = %d = (%s)_2, |I| = %d + %d = %d" % (n, m, binary(m), bits(n), bits(m), bits(n) + bits(m)))
a = [6, 1, 3, 8]
b = [bits(x) for x in a]
print("One array: a = %s, bits per entry = %s, |I| = %d" % (a, b, sum(b)))

# The bounds of the lecture notes (values rounded to two decimals)
print()
print("One number: log2(n) = %.2f <= |I| = %d <= 2 log2(n) = %.2f" % (log2(n), bits(n), 2 * log2(n)))
s = log2(n) + log2(m)
print("Two numbers: log2(n) + log2(m) = %.2f <= |I| = %d <= 2 (log2(n) + log2(m)) = %.2f" % (s, bits(n) + bits(m), 2 * s))
print("Two numbers and maximum: log2(max) = %.2f <= |I| = %d <= 4 log2(max) = %.2f" % (log2(max(n, m)), bits(n) + bits(m), 4 * log2(max(n, m))))
print("One array: n = %d <= |I| = %d <= n (1 + log2 M) = %.2f" % (len(a), sum(b), len(a) * (1 + log2(max(a)))))

# The upper bound is not tight: a[1] = M = 2^n, a[2..n] = 1 has |I| = 2n, but n(1 + log2 M) = n(n + 1)
print()
print("   n |      M |  |I| = 2n | n (1 + log2 M)")
for n in range(1, 11):
    M = 2 ** n
    a = [M] + [1] * (n - 1)
    print("%4d | %6d | %9d | %13d" % (n, M, sum(bits(x) for x in a), n * (n + 1)))
```

## 2. Sum of the entries of an array: SumArray

**Input:** an array $a[1..n]$ of $n \ge 1$ natural numbers. **Output:** $a[1]+a[2]+\cdots+a[n]$. Number of steps: $S(n) = 3n + 4$, whatever the values.

Pseudo-code: section 11.6 of the [chapter page](https://fabiofurini.github.io/algorithms/analysis-recursion/11-input-size/).

```python
def sum_array(a):
    n = len(a)
    s = 0; cnt[2] += 1
    i = 0; cnt[3] += 1                 # i := 1 in the pseudo-code
    cnt[4] += 1                        # the while condition is counted at every evaluation
    while i < n:                       # i <= n in the pseudo-code
        s = s + a[i]; cnt[5] += 1
        i = i + 1; cnt[6] += 1
        cnt[4] += 1
    cnt[8] += 1
    return s
```

```python
a = [6, 1, 3, 8]
n = len(a)
reset()
s = sum_array(a)
print("a = %s, n = %d, sum = %d" % (a, n, s))
print_steps(2, 3, 4, 5, 6, 8)
print("Formula: S(n) = 3n + 4 = %d" % (3 * n + 4))

# Worst case: the array of |I| ones has input size |I| and takes exactly 3|I| + 4 steps
print()
print("Array of |I| ones:")
for size in [4, 8, 16, 32]:
    reset()
    sum_array([1] * size)
    print("|I| = %2d: S = %3d = 3|I| + 4 = %d" % (size, sum(cnt), 3 * size + 4))
```
